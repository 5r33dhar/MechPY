```python
# Import the main Tkinter library for building the Graphical User Interface (GUI)
import tkinter as tk  
 # Import ttk for modern styling options and messagebox for popup alert windows
from tkinter import ttk, messagebox 
# Import NumPy for high-performance, fast vector-based mathematical arrays
import numpy as np  
# Import PyPlot from Matplotlib to create and format data visualization graphs
import matplotlib.pyplot as plt 
 # Import the special backend engine to embed Matplotlib charts inside a Tkinter window
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg 



# ==========================================
# 1. INITIALIZE & CONFIGURE THE GUI WINDOW
# ==========================================
# Create the primary application main window instance
beam_win = tk.Tk()  
 # Set the title text that appears at the top bar of the application window
beam_win.title("SFD and BMD Calculator: Simply Supported Beam") 

# Define a uniform horizontal layout spacing variable (in pixels) for padding out components
padx = 10  
# Define a uniform vertical layout spacing variable (in pixels) for padding out components
pady = 10  

# --- Setup Input Fields for Beam Length ---
# Create a text label widget explaining the beam length input field
label_length = tk.Label(beam_win, text="Beam Length BL (m):")  
 # Assign the label to cell row 0, column 0 inside the window grid grid
label_length.grid(row=0, column=0, padx=padx, pady=pady) 
# Create an interactive text input text box where users can type the total length of the beam
Entry_length = tk.Entry(beam_win)  
# Place the text box into cell row 0, column 1 right next to its label
Entry_length.grid(row=0, column=1, padx=padx, pady=pady)  

# --- Setup Input Fields for Load Magnitude ---
# Create a text label widget explaining the force magnitude input field
label_Point_load = tk.Label(beam_win, text="Load Magnitude LM (N):")  
# Place this descriptive text label into cell row 1, column 0 of the window grid
label_Point_load.grid(row=1, column=0, padx=padx, pady=pady)  
# Create an interactive text input text box where users can type the applied force value in Newtons
Point_load = tk.Entry(beam_win)  
# Position this load text box into cell row 1, column 1 of the grid
Point_load.grid(row=1, column=1, padx=padx, pady=pady)  

# --- Setup Input Fields for Load Distance ---
# Create a label indicating where the point load sits along the beam axis
label_Distance = tk.Label(beam_win, text="Distance from Left Support LA (m):")  
# Position this position label layout into cell row 2, column 0 of the grid
label_Distance.grid(row=2, column=0, padx=padx, pady=pady)  
# Create an interactive text entry box for entering the distance from the origin (left support)
Entry_Distance = tk.Entry(beam_win)  
# Align this distance entry input block into cell row 2, column 1 of the layout
Entry_Distance.grid(row=2, column=1, padx=padx, pady=pady)  



# ==========================================
# 2. EMBED MATPLOTLIB GRAPHING CANVAS
# ==========================================

# Initialize a blank canvas figure layout with specific physical window sizes (8 inches wide, 6 inches tall)
fig = plt.figure(figsize=(8, 6))  
# Slice the figure layout into grid blocks and create the upper plot axis slot (2 rows, 1 column, index 1) for SFD
ax1 = fig.add_subplot(211)  
# Slice the figure layout into grid blocks and create the lower plot axis slot (2 rows, 1 column, index 2) for BMD
ax2 = fig.add_subplot(212)  
# Automatically reposition and spaces titles and labels to prevent text overlaps between the two subplots
fig.tight_layout(pad=3.0)  

# Wrap the Matplotlib active window element structure into a compatible Tkinter component format
canvas = FigureCanvasTkAgg(fig, master=beam_win)  
# Place the structural graphics window spanning across rows 0 and 1 down inside cell row 5
canvas.get_tk_widget().grid(row=5, column=0, columnspan=2, padx=padx, pady=pady)  


# ==========================================
# 3. CORE CALCULATION & PLOTTING LOGIC
# ==========================================

# Define the main function callback triggered when users hit the execution calculation command action button
def calculate_sfd_bmd():  
    # Extract raw text string string from length box and safely parse convert it into a numeric floating-point decimal value
    lengthBeam = float(Entry_length.get())  
    # Extract text content from load value box and convert it into a numeric floating-point decimal value 
    load = float(Point_load.get())  
    # Extract text coordinates from distance box and convert it into a numeric floating-point decimal value
    LengthA = float(Entry_Distance.get())  


    # --- Structural Statics Math Foundations ---
    # Apply statics moment equilibrium formulas to compute vertical reaction upward forces at the left support (A)
    Ra = load * (1 - (LengthA / lengthBeam))  
    # Solve basic global force balance check systems (Sum of Fy = 0) 
    # to extract structural reaction force remaining at the right support (B)
    Rb = load - Ra 


    # --- High-Speed Vector Engineering Generation ---
    # Initialize a continuous array timeline sequence tracking 
    # points from 0 to length limits in tiny 10mm data step updates
    x = np.arange(0, lengthBeam + 0.01, 0.01)  

    
    """Run lighting-fast array processing operations across all x steps using vector condition 
    evaluations to avoid slow code loops"""
    """Evaluates position points: assigns positive reaction Ra before load position, 
    and drops by load value after it passes the load location, creating a stepwise shear force profile"""
    ShearForce = np.where(x < LengthA, Ra, Ra - load) 
    # Tracks linear bending moments: increases linearly up to the load position point, then decreases towards zero   
    MomentB = np.where(x < LengthA, Ra * x, Ra * x - load * (x - LengthA))  

    # --- Reset Subplots Before Redrawing ---
    """Clear previous data plots, lines, grids, titles, 
    and legends completely out from the upper Shear Force display panel space# """
    ax1.clear()  
    """ # Clear old datasets, drawings, markers, 
    and axis adjustments out from the lower Bending Moment display section area# """
    ax2.clear()   



    # --- Draw the Shear Force Diagram (SFD) ---
    ax1.step(x, ShearForce, where="post", color="blue", linewidth=2)  # Render a step chart line capturing sudden shear drops exactly where point forces sit
    ax1.axhline(0, color='black', lw=1, linestyle='--')  # Superimpose a clean horizontal reference datum baseline tracking where cross sections hit exactly 0 Newtons
    ax1.set_title('Shear Force Diagram (SFD)')  # Apply text formatting properties marking the upper plot section banner name
    ax1.set_xlabel('Position along the beam (m)')  # Apply text structural labels directly beneath horizontal structural lengths coordinate markers
    ax1.set_ylabel('Shear Force (N)')  # Set vertical data metrics labeling force values on the left Y-axis profile grid tracking lines
    ax1.grid(True)  # Draw standard graph checkerboards lines across the subplot area to track geometric coordinates easily

    # --- Draw the Bending Moment Diagram (BMD) ---
    ax2.plot(x, MomentB, color='red', linewidth=2)  # Trace smooth linear continuous paths matching calculated load trends across all coordinate variations
    ax2.axhline(0, color='black', lw=1, linestyle='--')  # Superimpose a reference line marker defining bending states passing back through 0 N-m balances
    ax2.set_title('Bending Moment Diagram (BMD)')  # Set the title at the top of the lower subplot area
    ax2.set_xlabel('Position along the beam (m)')  # Standardize layout tracking markers directly under structural length markers
    ax2.set_ylabel('Bending Moment (N-m)')  # Print target measurement dimensions tracking structural bending moments along the left side
    ax2.grid(True)  # Superimpose grid checkerboard patterns to aid in checking extreme value bounds across coordinates


    # --- Force Graphic Updates ---
    canvas.draw()  # Refresh the layout to render the updated charts in the UI viewport


# ==========================================
# 4. ACTION BUTTONS & MAINLOOP ACTIVATION
# ==========================================
# Construct the core button that activates calculations on click, with green background and white text colors
plot_button = tk.Button(beam_win, text="Calculate SFD and BMD", command=calculate_sfd_bmd, bg="#4CAF50", fg="white")  
# Place button onto grid row 4, stretched out across both horizontal columns evenly
plot_button.grid(row=4, column=0, columnspan=2, padx=padx, pady=pady)  


# Pre-render a blank, empty chart coordinate system grid canvas right away upon loading the application layout structure
canvas.draw()  
# Keep the Tkinter application active and running to handle clicks, window scaling, and text entries until closed
beam_win.mainloop()  
```

```python


# create gui window
beam_win = tk.Tk() # create a new Tkinter window
beam_win.title("SFD and BMD Calculator: Simply Supported Beam") # set the title of the window

padx = 10 # set the horizontal padding for widgets
pady = 10 # set the vertical padding for widgets

# label for beam length
label_length = tk.Label(beam_win, text="Beam Length BL (m):")
label_length.grid(row=0, column=0, padx=padx, pady=pady)
Entry_length = tk.Entry(beam_win)
Entry_length.grid(row=0, column=1, padx=padx, pady=pady)

# label for load magnitude
label_Point_load = tk.Label(beam_win, text="Load Magnitude LM (N):")
label_Point_load.grid(row=1, column=0, padx=padx, pady=pady)
Point_load = tk.Entry(beam_win)
Point_load.grid(row=1, column=1, padx=padx, pady=pady)

# label for distance of the point load from the left support
label_Distance = tk.Label(beam_win, text="Distance from Left Support LA (m):")
label_Distance.grid(row=2, column=0, padx=padx, pady=pady)
Entry_Distance = tk.Entry(beam_win)
Entry_Distance.grid(row=2, column=1, padx=padx, pady=pady)

fig = plt.figure(figsize=(8, 6))
ax1 = fig.add_subplot(211)
ax2 = fig.add_subplot(212)
canvas = FigureCanvasTkAgg(fig, master=beam_win)
canvas.get_tk_widget().grid(row=5, column=0, columnspan=2, padx=padx, pady=pady)


def calculate_sfd_bmd():
    try:
        lengthBeam = float(Entry_length.get())
        load = float(Point_load.get())
        LengthA = float(Entry_Distance.get())
    except ValueError:
        messagebox.showerror("Invalid input", "Enter numeric values for all three inputs.")
        return

    if not np.isfinite([lengthBeam, load, LengthA]).all():
        messagebox.showerror("Invalid input", "All inputs must be finite numbers.")
        return
    if lengthBeam <= 0:
        messagebox.showerror("Invalid input", "Beam length must be greater than zero.")
        return
    if not 0 <= LengthA <= lengthBeam:
        messagebox.showerror("Invalid input", "Load distance must be between 0 and the beam length.")
        return

    Ra = load * (1 - LengthA / lengthBeam)
    Rb = load - Ra

    sample_count = max(2, int(np.ceil(lengthBeam / 0.01)) + 1)
    Point_lengthBeam = np.linspace(0, lengthBeam, sample_count)
    ShearForce = np.where(Point_lengthBeam < LengthA, Ra, Ra - load)
    MomentB = Ra * Point_lengthBeam - load * np.maximum(Point_lengthBeam - LengthA, 0)

    ax1.clear()
    ax1.step(Point_lengthBeam, ShearForce, where="post", color="blue")
    ax1.set_title("Shear Force Diagram")
    ax1.set_ylabel("Shear Force (N)")
    ax1.grid(True)

    ax2.clear()
    ax2.plot(Point_lengthBeam, MomentB, color="red")
    ax2.set_title("Bending Moment Diagram")
    ax2.set_xlabel("Position along the beam (m)")
    ax2.set_ylabel("Bending Moment (N m)")
    ax2.grid(True)

    fig.tight_layout()
    canvas.draw()
    return Point_lengthBeam, ShearForce, MomentB


plot_button = tk.Button(beam_win, text="Calculate SFD and BMD", command=calculate_sfd_bmd)
plot_button.grid(row=4, column=0, columnspan=2, padx=padx, pady=pady)

# display gui window
beam_win.mainloop() # start the Tkinter event loop
```

```python
import tkinter as tk
from tkinter import ttk, messagebox
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg

# create gui window
beam_win = tk.Tk()
beam_win.title("SFD and BMD Calculator: Simply Supported Beam")

pad_x = 0
pad_y = 2.5

# label for beam length
l_length = ttk.Label(beam_win, text="Length of the Beam L (m):")
l_length.grid(row=0, column=0, padx=pad_x, pady=pad_y)
entry_L = ttk.Entry(beam_win)
entry_L.grid(row=0, column=1, padx=pad_x, pady=pad_y)

# label for Point Load on the beam
l_load = ttk.Label(beam_win, text="Value of the Point Load P (N):")
l_load.grid(row=1, column=0, padx=pad_x, pady=pad_y)
entry_load = ttk.Entry(beam_win)
entry_load.grid(row=1, column=1, padx=pad_x, pady=pad_y)

# label for Distance of the Point Load From Left Side
l_la = ttk.Label(beam_win, text="Distance P from Left End la (m):")
l_la.grid(row=2, column=0, padx=pad_x, pady=pad_y)
entry_la = ttk.Entry(beam_win)
entry_la.grid(row=2, column=1, padx=pad_x, pady=pad_y)

# ── ENHANCEMENT 2: Output labels for Ra and Rb ──────────────────────────────
l_ra_title = ttk.Label(beam_win, text="Reaction at A - Ra (N):")
l_ra_title.grid(row=3, column=0, padx=pad_x, pady=pad_y)
l_ra_value = ttk.Label(beam_win, text="—", foreground="blue")
l_ra_value.grid(row=3, column=1, padx=pad_x, pady=pad_y)

l_rb_title = ttk.Label(beam_win, text="Reaction at B - Rb (N):")
l_rb_title.grid(row=4, column=0, padx=pad_x, pady=pad_y)
l_rb_value = ttk.Label(beam_win, text="—", foreground="blue")
l_rb_value.grid(row=4, column=1, padx=pad_x, pady=pad_y)
# ────────────────────────────────────────────────────────────────────────────





def sfd_bmd():
    # ── ENHANCEMENT 1: Input validation ─────────────────────────────────────
    try:
        L  = float(entry_L.get())
        la = float(entry_la.get())
        P  = float(entry_load.get())
    except ValueError:
        messagebox.showerror("Invalid Input", "Please enter valid numbers in all fields.")
        return

    if la >= L:
        messagebox.showerror("Invalid Input", "Distance 'la' must be less than beam length 'L'.")
        return

    if L <= 0 or P <= 0:
        messagebox.showerror("Invalid Input", "Beam length and load must be greater than zero.")
        return
    # ────────────────────────────────────────────────────────────────────────

    Ra = P * (1 - (la / L))   # reaction at left support
    Rb = P - Ra                # reaction at right support

    # ── ENHANCEMENT 2: Update Ra / Rb labels ────────────────────────────────
    l_ra_value.config(text=f"{Ra:.2f} N")
    l_rb_value.config(text=f"{Rb:.2f} N")
    # ────────────────────────────────────────────────────────────────────────

    SF = []
    Lb = []
    Mb = []

    for x in np.arange(0, L + 0.01, 0.01):
        if x < la:
            sf = Ra
            m  = Ra * x
        else:
            sf = Ra - P
            m  = (Ra * x) - (P * (x - la))
        Lb.append(x)
        SF.append(sf)
        Mb.append(m)

    fig.clear()

    # plot shear force
    ax1 = fig.add_subplot(211)
    ax1.plot(Lb, SF, color='blue')
    ax1.fill_between(Lb, SF, color='lightblue', alpha=0.5, hatch='//')
    ax1.axhline(0, color='black', lw=0.5)
    ax1.set_title("Shear Force Diagram")
    ax1.set_ylabel("Shear Force (N)")
    ax1.grid(True, linestyle='--', alpha=0.5)          # ENHANCEMENT 6: Gridlines

    # plot bending moment
    ax2 = fig.add_subplot(212)
    ax2.plot(Lb, Mb, color='green')
    ax2.fill_between(Lb, Mb, color='lightgreen', alpha=0.5, hatch='//')
    ax2.axhline(0, color='black', lw=0.5)
    ax2.set_title("Bending Moment Diagram")
    ax2.set_xlabel("Beam Length (m)")
    ax2.set_ylabel("Bending Moment (N·m)")              # ENHANCEMENT 3: Fixed label
    ax2.grid(True, linestyle='--', alpha=0.5)           # ENHANCEMENT 6: Gridlines

    # ── ENHANCEMENT 5: Annotate peak bending moment ──────────────────────────
    max_m       = max(Mb)
    max_m_index = Mb.index(max_m)
    max_m_x     = Lb[max_m_index]

    ax2.annotate(
        f"Max = {max_m:.2f} N·m",
        xy=(max_m_x, max_m),
        xytext=(max_m_x + 0.05 * L, max_m * 0.85),
        arrowprops=dict(arrowstyle='->', color='darkgreen'),
        fontsize=9,
        color='darkgreen'
    )
    # ────────────────────────────────────────────────────────────────────────

    fig.subplots_adjust(left=0.2, hspace=0.3)
    canvas.draw()


def clear_plot():
    # ── ENHANCEMENT 4: Clear / Reset ────────────────────────────────────────
    entry_L.delete(0, tk.END)
    entry_load.delete(0, tk.END)
    entry_la.delete(0, tk.END)
    l_ra_value.config(text="—")
    l_rb_value.config(text="—")
    fig.clear()
    canvas.draw()
    # ────────────────────────────────────────────────────────────────────────


# Buttons
btn_frame = ttk.Frame(beam_win)
btn_frame.grid(row=5, column=0, columnspan=2, pady=10)

plot_button  = ttk.Button(btn_frame, text="Plot SFD and BMD", command=sfd_bmd)
plot_button.grid(row=0, column=0, padx=5)

clear_button = ttk.Button(btn_frame, text="Clear / Reset", command=clear_plot)  # ENHANCEMENT 4
clear_button.grid(row=0, column=1, padx=5)

fig    = plt.Figure(figsize=(8, 6), dpi=100)
canvas = FigureCanvasTkAgg(fig, beam_win)
canvas.get_tk_widget().grid(row=6, column=0, columnspan=2)


# display gui window
beam_win.mainloop()
```
