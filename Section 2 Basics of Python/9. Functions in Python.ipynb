```python
# Area of cylinder
def area(d, h):
    A = 3.14 * d * h
    return A

area(1,5)
```

```python
# Area of rectangle
def rectangle_area(b,w):
    return b*w

print(rectangle_area(10,10))
```

```python
def four_bar(li,coup,lo):
    if (li+lo) > coup:
        print("This mechanism is possible")
    else:
        print("This is locked mechanism")
        
four_bar(10,18,10)
```
