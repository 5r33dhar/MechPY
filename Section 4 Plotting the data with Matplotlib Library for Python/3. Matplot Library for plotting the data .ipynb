```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.gridspec as gsp



x = np.arange(0,5,0.1)
y1 = 2*x ** 2
y2 = 3*x ** 2 + 2*x

y3 = np.sin(x)
y4 = np.cos(x)


fig = plt.figure(figsize = (8,6),facecolor=[0.5,0.5,0.5])
gs = gsp.GridSpec(2,2)

ax1 = fig.add_subplot(gs[0,: ],facecolor=[0.0,0.5,0.5])
ax1.plot(x,y1,label="y1_data", color='red')
ax1.set_title("$y_1$=$2x^2$")
ax1.set_xlim(0,5)
ax1.set_ylim(0,50)
ax1.grid()
ax1.legend()

ax2=fig.add_subplot(gs[1,0])
ax2.plot(x,y2,label="y2_data")
ax2.set_title("$y_2$=$3x^2+2*x$")
ax2.grid(True)
ax2.legend()

ax3=fig.add_subplot(gs[1,1])
ax3.plot(x,y3,label="y3_data")
ax3.set_title("$y_3$=$sin(x)$")
ax3.grid()
ax3.legend()

plt.tight_layout()
```

## Polar Plot 

```python
a= 8
b = 5
theta = np.arange(0,2*np.pi,np.pi/180)

for t in theta:
    r = (a*b)/np.sqrt((a*np.sin(t)) ** 2+(b*np.cos(t)) ** 2)
plt.polar(t,r,'b.')

plt.show()
```
