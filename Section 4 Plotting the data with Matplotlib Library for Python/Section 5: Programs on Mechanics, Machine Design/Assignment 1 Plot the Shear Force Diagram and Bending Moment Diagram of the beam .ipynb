```python
import numpy as np
import matplotlib.pyplot as plt

# Inputs (using SI units: N, m)
W = 4000  # UDL in N/m
F = 2000  # Point load at A in N
L = 4.0   # Total length of the beam in m

# Segment positions from A (x = 0)
a = 1.0   # Point B
b = 3.0   # Point C

# Generate x values along the beam
x_vals = np.linspace(0, L, 400)

SF = []
M = []

# Evaluate Shear Force and Bending Moment at each point x
for x in x_vals:
    if x < a:
        # Region AB (0 to 1m)
        sf = -F
        m = -F * x
    elif a <= x < b:
        # Region BC (1m to 3m)
        sf = -F - W * (x - a)
        m = -F * x - 0.5 * W * (x - a)**2
    else:
        # Region CD (3m to 4m)
        sf = -F - W * (b - a)
        m = -F * x - W * (b - a) * (x - (a + b)/2)
        
    SF.append(sf)
    M.append(m)

# Plotting the diagrams
plt.figure(figsize=(8, 7))

# 1. Shear Force Diagram
plt.subplot(2, 1, 1)
plt.plot(x_vals, SF, color='green', linewidth=2)
plt.fill_between(x_vals, SF, color="green", hatch="||", alpha=0.3)
plt.title("Shear Force Diagram (SFD)", fontsize=12, fontweight='bold')
plt.xlabel("Length of Beam (m)")
plt.ylabel("Shear Force (N)")
plt.xlim(0, L)
plt.grid(True, linestyle='--', alpha=0.6)

# 2. Bending Moment Diagram
plt.subplot(2, 1, 2)
plt.plot(x_vals, M, color='red', linewidth=2)
plt.fill_between(x_vals, M, color='red', hatch="\\", alpha=0.3)
plt.title("Bending Moment Diagram (BMD)", fontsize=12, fontweight='bold')
plt.xlabel("Length of Beam (m)")
plt.ylabel("Moment (Nm)")
plt.xlim(0, L)
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout(pad=3.0)
plt.show()
```
