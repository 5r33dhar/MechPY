```python
# # Single degree of Freedom system- 

import numpy as np
import matplotlib.pyplot as plt
from numpy.linalg import inv
from  numpy import sin ,cos, pi 

# Inputs 

M = 100  # Mass
C = 10  # damping
K = 50 # stiffness

w = 20

end_time = 40
time_step = 0.001

F0 = 100 #N

A = np.array([[M,0],
              [0,1]])

B = np.array([[C , K],
              [-1, 0]])

x = np.array([[2],
              [1]])

F= np.zeros([2,1])

# empty arrays
S = []
V = []
T = []

time = np.arange(0,end_time,time_step)

for t in time:
    F[0] = F0*sin(w*t)
    X = x + (time_step*(inv(A)).dot(F-B.dot(x)))
    x = X
    
    S.append(X[1])
    V.append(X[0])
    T.append(t)


plt.plot(T,S)
plt.show()
```

```python
#AI

import numpy as np
import matplotlib.pyplot as plt
from numpy.linalg import inv
from numpy import sin

# System Variables
M = 100   # Mass
C = 10    # Damping
K = 50    # Stiffness

w = 20    # Forcing frequency
F0 = 100  # Force amplitude

end_time = 40
time_step = 0.0005 # Reduced time step for Euler stability
time = np.arange(0, end_time, time_step)

# Standard State-Space Matrices
# Vector layout: x = [[displacement], [velocity]]
A = np.array([[1, 0],
              [0, M]])

B = np.array([[0, -1],
              [K, C]])

# Initial conditions: displacement = 1, velocity = 2
x = np.array([[1.0],
              [2.0]])

F = np.zeros([2, 1])

# Storage arrays
displacement = []
velocity = []
time_log = []

A_inv = inv(A)

# Simulation Loop (Euler Forward Method)
for t in time:
    F[1] = F0 * sin(w * t)
    
    # State derivative calculation
    x_dot = A_inv.dot(F - B.dot(x))
    
    # Update state
    x = x + time_step * x_dot
    
    # Store results
    displacement.append(x[0][0])
    velocity.append(x[1][0])
    time_log.append(t)

# Plotting Results
plt.figure(figsize=(10, 5))
plt.plot(time_log, displacement, label='Displacement (x)')
plt.title('SDOF System Response')
plt.xlabel('Time (s)')
plt.ylabel('Displacement (m)')
plt.grid(True)
plt.legend()
plt.show()
```
