```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import fftpack

t = np.linspace(-10,10,200)
dt = t[1] - t[0]

freq = 50
width = 2

g - np.exp(-np.abs(t)/width) *np.sin(2*np.pi*freq*t)

G = fftpack.fft(g)
f = fftpack.fftfreq(g.size,d=dt)

plt.figure(figsize=(15,7.5))
plt.subplot(211)
plt.plot(t,g)

plt.subplot(212)
plt.plot(f,abs(G))
plt.show()
```

```python
#example  for  sinwave 

from scipy import fftpack
import numpy as np
import matplotlib.pyplot as plt

fs = 500
t = np.arange(0,1,1/fs)
f = 30
y= np.sin(2*np.pi*f*t) + np.cos(2*np.pi*f*t)

y_fft = fftpack.fft(y)

plt.figure(figsize=(10,10))
plt.subplot(311)
plt.plot(t,y)
plt.title('sin wave')

plt.subplot(312)
plt.plot(abs(y_fft))
plt.title('fft_curve')
plt.xlim(0,50)

plt.subplot(312)
plt.plot(np.real(y_fft))
plt.title('real fft_curve')
plt.xlim(0,50)

plt.subplot(322)
plt.plot(np.imag(y_fft))
plt.title('imag fft_curve')
plt.xlim(0,50)
```

```python
from scipy import fftpack
import numpy as np
import matplotlib.pyplot as plt
from scipy import fftpack
import matplotlib.pyplot as plt

width = 2
freq = 0.5
t = np.linspace(-10,10,200)
g = np.exp(-np.abs(t)/width) * np.sin(2*np.pi*freq*t)
dt = t[1]-t[0]

G = fftpack.fft(g)
f = fftpack.fftfreq(g.size, d=dt)

plt.figure(figsize=(10,10))
plt.subplot(211)
plt.plot(t,g)

plt.subplot(212)
plt.plot(f,np.imag(G),label = 'real')
#plt.yscale('log')

plt.legend()

plt.show()
```
