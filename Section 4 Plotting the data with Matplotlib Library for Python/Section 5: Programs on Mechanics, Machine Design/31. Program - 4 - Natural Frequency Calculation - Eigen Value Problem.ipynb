```python
 
# Natural Frequencies of system

import numpy as np
from scipy import linalg

m1,m2,m3 = (5,5,5)
k1,k2,k3 = (100,100,100)

M = np.array ([[m1,0,0],
               [0,m2,0],
               [0,0,m3]])

K = np.array ([[k1+k2,-k2,0],
               [-k2,k2+k3,-k3],
               [0,-k3,k3]])


M

K

u,v = linalg.eigh(K,M)

print(u) # eigen value
print ("")
print (v) # eigen vector

nat_freq = np.sqrt(u)

print ("")

print ("Natural  freequency ",nat_freq)

print ("")

print("First natural frequency :",nat_freq[0])
print("Second natural frequency :",nat_freq[1])
print("Third natural frequency :",nat_freq[2])




```

```python
import numpy as np

#       A          b
#   1x  +   2y  =  3
#   3x  +   4y  = 17


# Coefficient matrix A
A = np.array([[1, 2], 
              [3, 4]])

# Constant vector b
b = np.array([3, 17])

# Solve for x and y (Ax = b)

solution = np.linalg.solve(A, b)

print("Solution [x, y]:", solution)
# Output: Solution [x, y]: [11. -4.]
```

```python
import numpy as np

A = np.array([[1, 2], 
              [3, 4]])

# Compute the determinant
det = np.linalg.det(A)
print("Determinant:", det) 
# Output: Determinant: -2.0000000000000004

# Compute the multiplicative inverse
inverse = np.linalg.inv(A)
print("Inverse Matrix:\n", inverse)
# Output:
# [[-2.   1. ]
#  [ 1.5 -0.5]]
```

```python
import numpy as np

# Matrix Multiplication (Dot Product)
#You can multiply two matrices or vectors using the @ operator or np.dot()

A = np.array([[1, 2], 
              [3, 4]])

B = np.array([[5, 6], 
              [7, 8]])

# Matrix multiplication
result = A @ B

Dot =np.dot(A,B)

print(result)
print(Dot)
# Output:
# [[19 22]
#  [43 50]]
```
