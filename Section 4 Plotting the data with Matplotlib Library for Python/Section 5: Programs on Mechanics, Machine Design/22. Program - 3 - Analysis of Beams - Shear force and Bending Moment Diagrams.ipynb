```python
import numpy as np
import matplotlib.pyplot as plt

# inputs
L = 10 # m
W = 50 # N/m

# Reaction at A (Left )
R = W*L/2
x = np.linspace(0,L,100)

print(x)  #  points  on   Beam  with  x distance  from  left  

X = []
SF = []
M = []

for l in x:
    sf = R -(W*l)
    m = (R*l) - (W*l ** 2/2)

    X.append(l) # adding  value to array 
    SF.append(sf) # adding  value to array 
    M.append(m) # adding  value to array 


plt.figure(figsize=(7.5,5),dpi=100)
plt.subplot(2,1,1)
plt.plot(X,SF,'g')
plt.xlim(0,L)
plt.ylim(min(SF),max(SF))

plt.title("Shear Force Diagram")
plt.ylabel("Shear Force")
plt.tight_layout(pad=4.0)
plt.grid()

plt.subplot(2,1,2)
plt.plot(X,M,'r')
plt.xlim(0,L)
plt.ylim(min(M),max(M))

plt.title("Shear Force Diagram")
plt.ylabel("Shear Force")
plt.tight_layout(pad=4.0)
plt.grid()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# inputs
L = 10 # m
W = 50 # N/m

# Reaction at A (Left )
R = W*L/2
x = np.linspace(0,L,100)

print(x)  #  points  on   Beam  with  x distance  from  left  

X = []
SF = []
M = []

for l in x:
    sf = R -(W*l)
    m = (R*l) - (W*l ** 2/2)

    X.append(l)
    SF.append(sf)
    M.append(m)


plt.figure(figsize=(7.5,5),dpi=100)
plt.subplot(2,1,1)
plt.plot(X,SF,'g')
plt.xlim(0,L)
plt.ylim(min(SF),max(SF))
plt.fill_between(X,SF,color="green",alpha=0.5,hatch="||")
plt.title("Shear Force Diagram")
plt.ylabel("Shear Force")
plt.tight_layout(pad=4.0)
plt.grid()

plt.subplot(2,1,2)
plt.plot(X,M,'r')
plt.xlim(0,L)
plt.ylim(min(M),max(M))
plt.fill_between(X,M,color="red",alpha=0.2,hatch="||")
plt.title("Shear Force Diagram")
plt.ylabel("Shear Force")
plt.tight_layout(pad=4.0)
plt.grid()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg
import tkinter as tk
from tkinter import ttk

# --- Calculation and Update Function ---
def update_plots(*args):
    try:
        L = float(length_var.get())
        W = float(load_var.get())
    except ValueError:
        return

    # 1. Recalculate forces
    R = W * L / 2
    x = np.linspace(0, L, 100)
    
    SF = R - (W * x)
    M = (R * x) - (W * x ** 2 / 2)

    # 2. Find Max Absolute Values and locations
    max_sf = max(abs(SF))
    max_m = max(M)  # Max moment occurs at the center for a simply supported beam
    max_m_x = L / 2

    # Update Tkinter interface labels
    max_sf_label.config(text=f"Max Shear Force: {max_sf:.2f} N")
    max_m_label.config(text=f"Max Bending Moment: {max_m:.2f} N·m")

    # Clear previous drawings
    ax1.clear()
    ax2.clear()

    # --- Top Plot: Shear Force ---
    ax1.plot(x, SF, 'g', label='SF')
    ax1.fill_between(x, SF, color="green", alpha=0.5, hatch="||")
    
    # Annotate Max SF on plot (Occurs at x=0 and x=L)
    ax1.plot([0, L], [R, -R], 'go', markersize=6)
    ax1.annotate(f'Max: {max_sf:.1f} N', xy=(0, R), xytext=(L*0.1, R*0.7),
                 arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=6))

    ax1.set_xlim(0, L)
    ax1.set_ylim(min(SF) - 10, max(SF) + 10) 
    ax1.set_title("Shear Force Diagram")
    ax1.set_ylabel("Shear Force (N)")
    ax1.grid(True)

    # --- Bottom Plot: Bending Moment ---
    ax2.plot(x, M, 'r', label='BM')
    ax2.fill_between(x, M, color="red", alpha=0.2, hatch="||")
    
    # Annotate Max Bending Moment on plot (Occurs at mid-span)
    ax2.plot(max_m_x, max_m, 'ro', markersize=6)
    ax2.annotate(f'Max: {max_m:.1f} N·m', xy=(max_m_x, max_m), xytext=(max_m_x + L*0.05, max_m * 0.8),
                 arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=6))

    ax2.set_xlim(0, L)
    ax2.set_ylim(min(M) - 10, max(M) + 10)
    ax2.set_title("Bending Moment Diagram")
    ax2.set_ylabel("Bending Moment (N·m)")
    ax2.set_xlabel("Beam Length (m)")
    ax2.grid(True)

    # Redraw canvas
    fig.tight_layout(pad=2.0)
    canvas.draw()

# --- GUI Setup ---
root = tk.Tk()
root.title("Interactive Beam Analysis")

# Main container frames
control_frame = ttk.Frame(root, padding="10")
control_frame.pack(side=tk.TOP, fill=tk.X)

# New frame specifically to hold the text readouts cleanly
stats_frame = ttk.LabelFrame(root, text=" Calculated Maximum Values ", padding="10")
stats_frame.pack(side=tk.TOP, fill=tk.X, padx=10, pady=5)

plot_frame = ttk.Frame(root)
plot_frame.pack(side=tk.BOTTOM, fill=tk.BOTH, expand=True)

# Variables bound to sliders
length_var = tk.DoubleVar(value=10.0)
load_var = tk.DoubleVar(value=50.0)

length_var.trace_add("write", update_plots)
load_var.trace_add("write", update_plots)

# --- Controls UI ---
# Length (L) Slider
ttk.Label(control_frame, text="Beam Length (L):").grid(row=0, column=0, sticky="w", padx=5)
l_scale = ttk.Scale(control_frame, from_=1, to=30, variable=length_var, orient="horizontal", length=200)
l_scale.grid(row=0, column=1, padx=5)
ttk.Label(control_frame, textvariable=length_var).grid(row=0, column=2, padx=5)

# Load (W) Slider
ttk.Label(control_frame, text="UDL Load (W):").grid(row=1, column=0, sticky="w", padx=5)
w_scale = ttk.Scale(control_frame, from_=0, to=150, variable=load_var, orient="horizontal", length=200)
w_scale.grid(row=1, column=1, padx=5)
ttk.Label(control_frame, textvariable=load_var).grid(row=1, column=2, padx=5)

# --- Window Text Labels for Max Values ---
max_sf_label = ttk.Label(stats_frame, text="Max Shear Force: 0.00 N", font=('Arial', 10, 'bold'), foreground="green")
max_sf_label.pack(anchor="w", pady=2)

max_m_label = ttk.Label(stats_frame, text="Max Bending Moment: 0.00 N·m", font=('Arial', 10, 'bold'), foreground="red")
max_m_label.pack(anchor="w", pady=2)

# --- Matplotlib Figure Setup ---
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(7.5, 6), dpi=100)

# Embed figure in Tkinter window
canvas = FigureCanvasTkAgg(fig, master=plot_frame)
canvas.get_tk_widget().pack(fill=tk.BOTH, expand=True)

# Run initial calculation loop
update_plots()

# Start the application loop
root.mainloop()
```
