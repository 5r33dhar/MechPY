```python
import numpy as np
from scipy import linalg
import matplotlib.pyplot as plt
import cmath


m1, m2, m3 = 5, 5, 5
k1, k2, k3 = 1200, 1300, 1600

w= np.linspace(0,50,500)
alpha = 0.01
beta = 0.01

real =0
imag =1

I =complex(real,imag)

print(I)

X1=[]
X1a =[]  # response  of Mass 1
X2a =[]  # response  of Mass 2
X3a =[]  # response  of Mass 3
W=[]


for  f in w:
    M = np.array ([[m1,0,0],
                    [0,m2,0],
                    [0,0,m3]])

    K = np.array ([[k1+k2,-k2,0],
                    [-k2,k2+k3,-k3],
                    [0,-k3,k3]])


    C = (alpha*M) + (beta*K)

    Force = np.array([[50],
                     [0],
                     [0]])

    
    D= K-(f**2*M)-(I*f*C)  # Equation2 f is  w

    x =np.linalg.inv(D).dot(Force)
    X1.append(x[0])
    X1a.append(abs(x[0]))
    X2a.append(abs(x[1]))
    X3a.append(abs(x[2]))

    W.append(f) # Equation2 f is  w
print(x)
print(X1)
print(X1a)

plt.figure(figsize=(8,4))
plt.plot(w,X1a,label='Mass1')
plt.plot(w,X2a,label='Mass2')
plt.plot(w,X3a,label='Mass3')
plt.grid( which='both', color='0.5', linestyle='--')
plt.yscale('log')
plt.title("Frequency Response of MDOF",fontsize = 20)
plt.xlabel("Frequency Hz",fontsize = 15)
plt.ylabel("Amplitude",fontsize = 15)
plt.legend()
plt.show()

```

```python
# Frequency Response of MDOF  #given


import numpy as np 
import matplotlib.pyplot as plt 
from scipy import linalg
from numpy.linalg import inv
import cmath

m0, m1, m2 = (1, 1, 1)
k0, k1, k2 = (1600, 1600, 1600)
alpha, beta = 0.001,0.001

c1 = 10
c2 = 50
c3 = 10
c4 = 2

x = 0
y = 1
z = complex(x,y)

w = np.arange(0,100,0.1)
x1 = []
x2 = []
x3 = []
W = []
for i in w:
    M = np.array([[m0, 0, 0],
                  [0, m1, 0],
                  [0, 0, m2]])
    K = np.array([[k0+k1, -k1,   0],
                  [-k1, k1+k2, -k2],
                  [0,     -k2,  k2]])
    C = alpha*M + beta*K
    F = np.array([[10],
                  [0],
                  [0]])
 
    D = np.array(K- i**2*M - i*z*C)
    B = inv(D).dot(F)
    x1.append(abs(B[0]))
    x2.append(abs(B[1]))
    x3.append(abs(B[2]))
    W.append(i)

u,v = linalg.eigh(K,M)
nat_fre = np.sqrt(u)
print(nat_fre)

plt.figure(figsize=(20,10))


plt.subplot(2,2,1)
plt.plot(W,x1,label = 'dof_1')
plt.legend(loc = "best",fontsize = 15)
plt.yscale("log")
#plt.grid(b=True, which='both', color='0.5', linestyle='--')
plt.grid(which='both', color='0.5', linestyle='--')
#plt.ticklabel_format(style = 'sci',scilimits = (1,1),axis = 'y')
plt.xlim(W[0],W[-1])
#plt.ylim(1e-2,10.5)
plt.title("Frequency Response of MDOF",fontsize = 20)
plt.xlabel("Frequency Hz",fontsize = 15)
plt.ylabel("Amplitude",fontsize = 15)
plt.tight_layout(pad=3)

plt.subplot(2,2,2)
plt.plot(W,x2,label = 'dof_2')
plt.legend(loc = "best",fontsize = 15)
plt.yscale("log")
#plt.grid(b=True, which='both', color='0.5', linestyle='--')
plt.grid(which='both', color='0.5', linestyle='--')
#plt.ticklabel_format(style = 'sci',scilimits = (1,1),axis = 'y')
plt.xlim(W[0],W[-1])
#plt.ylim(1e-2,10.5)
plt.title("Frequency Response of MDOF",fontsize = 20)
plt.xlabel("Frequency Hz",fontsize = 15)
plt.ylabel("Amplitude",fontsize = 15)

plt.tight_layout(pad=3)

plt.subplot(2,2,3)
plt.plot(W,x3,label = 'dof_3')
plt.legend(loc = "best",fontsize = 15)
plt.yscale("log")
#plt.grid(b=True, which='both', color='0.5', linestyle='--')
plt.grid(which='both', color='0.5', linestyle='--')
#plt.ticklabel_format(style = 'sci',scilimits = (1,1),axis = 'y')
plt.xlim(W[0],W[-1])
#plt.ylim(1e-2,10.5)
plt.title("Frequency Response of MDOF",fontsize = 20)
plt.xlabel("Frequency Hz",fontsize = 15)
plt.ylabel("Amplitude",fontsize = 15)
plt.show()


```
