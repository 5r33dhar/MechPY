```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint


# soln = odeint(function,x0,x,arg)

def my_eq(y,x):
    dydx = x ** 3 + (2*x ** 2)

    return dydx

def In_eq(iy,x):
        iy = (x**4)/4 + (2*x**3)/3
        return iy


y0 = 4 # value of the y when x=0
x = np.linspace(0,1,20)

y = odeint(my_eq,y0,x)

y

iy = 4 # value of the y when x=0
yi = odeint(In_eq,iy,x)

plt.plot(x,y)
plt.plot(x,yi)
plt.show()
```
