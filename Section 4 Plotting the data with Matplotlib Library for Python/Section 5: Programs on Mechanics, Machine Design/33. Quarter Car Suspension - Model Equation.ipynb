```python
import numpy as np
import matplotlib.pyplot as plt
from numpy import sin, cos, pi

# ==========================================
# 1. PHYSICAL PARAMETERS (Quarter-Car Model)
# ==========================================
m1 = 400   # Mass of the vehicle body (sprung mass) in kg
m2 = 30    # Mass of the wheel/axle assembly (unsprung mass) in kg

k1 = 10e3  # Stiffness of the suspension spring in N/m
k2 = 40e3  # Stiffness of the tire in N/m

c1 = 1000  # Damping coefficient of the suspension shock absorber in N-s/m
c2 = 0     # Damping coefficient of the tire (assumed negligible)

# ==========================================
# 2. ROAD BUMP INPUT PARAMETERS
# ==========================================
a = 0.2   # Amplitude (height) of the road bump in meters
v = 0.85  # Forward velocity of the vehicle in m/s
d = 0.8   # Width of the road bump in meters

# Excitation frequency calculated from speed and bump width
w = 2*pi*(v/d) 

# ==========================================
# 3. SIMULATION TIME SETUP
# ==========================================
ts = 2.5       # Time at which the vehicle hits the bump (seconds)
t_step = 0.01  # Time step size for Euler integration (dt)
T = np.arange(0, 20, t_step) # Simulation time array from 0 to 20 seconds

# ==========================================
# 4. STATE-SPACE FORMULATION (dx/dt = A*x + F)
# ==========================================
A = np.array([[0.0, 1.0, 0.0, 0.0],
              [-k1/m1, -c1/m1, k1/m1, c1/m1],
              [0.0, 0.0, 0.0, 1.0],
              [k1/m2, c1/m2, -(k1+k2)/m2, -(c1+c2)/m2]])

# Initialize force vector and initial states (car starts at rest)
F = np.array([0.0, 0.0, 0.0, 0.0])
x = np.array([0.0, 0.0, 0.0, 0.0])

# Empty lists to record time-history data for plotting
d1 = []        # Vehicle displacement
v1 = []        # Vehicle velocity
d2 = []        # Wheel displacement
v2 = []        # Wheel velocity
road_prof = [] # Actual height of the road surface over time

# ==========================================
# 5. SIMULATION LOOP (Euler Integration)
# ==========================================
for t in T:
    # Check if the wheel is currently climbing/descending the bump
    if ts < t < (ts + d/v):
        # Calculate the geometric shape of the bump (0.5 * a * (1 - cos(...)))
        #Half  Wave  (0.5) +ve
        z_r = 0.5 * a * (1 - cos(w*(t-ts)))
        # Apply the tire-stiffness force to the wheel mass matrix element
        F[3] = (k2/m2) * z_r
    else:
        # Flat road profile before and after the bump
        z_r = 0.0
        F[3] = 0.0
    
    # State update using forward Euler method: x_new = x_old + dt * dx/dt
    x = x + t_step * (A.dot(x) + F)
    
    # Store calculated states and road geometry for this time step
    d1.append(x[0])
    v1.append(x[1])
    d2.append(x[2])
    v2.append(x[3])
    road_prof.append(z_r)

# ==========================================
# 6. DATA VISUALIZATION
# ==========================================
plt.figure(figsize=(10, 5), dpi=100)

# Plotting the three profiles
plt.plot(T, road_prof, label='Road Profile (Bump)', color='gray', linestyle=':', linewidth=2)
plt.plot(T, d2, label='Wheel Displacement', color='orange', alpha=0.8)
plt.plot(T, d1, label='Vehicle Body Displacement', color='blue', linewidth=2)

# Formatting plot elements
plt.grid(linestyle='--', color='k', alpha=0.3)
plt.legend(loc='best')
plt.xlim(T[0], T[-1])
plt.xlabel('Time (s)')
plt.ylabel('Displacement / Height (m)')
plt.title('Quarter-Car System Dynamics with Road Profile')

# Render the graph
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from numpy import sin, cos, pi
from matplotlib.widgets import Slider, Button, RadioButtons

# =============================================================================
# 1. VEHICLE PARAMETERS (Quarter-Car Model Constants)
# =============================================================================
m1 = 400     # Sprung mass (kg): The mass of the vehicle body/chassis
m2 = 30      # Unsprung mass (kg): The mass of the wheel, tire, and suspension links

k1 = 10000   # Suspension stiffness (N/m): The primary spring rate
k2 = 40000   # Tire stiffness (N/m): The vertical elasticity of the tire itself

c1 = 1000    # Suspension damping (N-s/m): Shock absorber damping coefficient
c2 = 0       # Tire damping (N-s/m): Inherent tire material damping (usually negligible)

# =============================================================================
# 2. ROAD PROFILE / EXCITATION INITIAL CONDITIONS
# =============================================================================
a = 0.2      # Peak road bump amplitude (meters)            
v = 0.85     # Constant forward velocity of the vehicle (m/s)
d = 0.8      # Structural physical width of the bump (meters)

# Wavelength frequency parameter for reference
w = 2 * pi * (v / d)

# F0 scales the input road profile into an acceleration forcing term.
# It converts road height (a) through tire stiffness (k2) relative to wheel mass (m2).
F0 = k2 * a / m2

# =============================================================================
# 3. TIME DOMAIN DISCRETIZATION
# =============================================================================
t_step = 0.01                     # Integration time step (10 milliseconds)
T = np.arange(0, 20 + t_step, t_step)  # Time array tracking from 0.0s to 20.0s

# =============================================================================
# 4. STATE-SPACE FORMULATION (Dynamic Matrix)
# =============================================================================
# The state vector 'y' represents: [y[0], y[1], y[2], y[3]]
# Where:
#   y[0] = Car Displacement (zs)
#   y[1] = Car Velocity (zs_dot)
#   y[2] = Wheel Displacement (zu)
#   y[3] = Wheel Velocity (zu_dot)
A = np.array([[0.0, 1.0, 0.0, 0.0],
              [-k1/m1, -c1/m1, k1/m1, c1/m1],
              [0.0, 0.0, 0.0, 1.0],
              [k1/m2, c1/m2, -(k1+k2)/m2, -(c1+c2)/m2]])

# Initialize the force vector [0, 0, 0, Wheel_Acceleration_Forcing]
F = np.array([0.0, 0.0, 0.0, 0.0])

# Initialize the state vector at equilibrium (all zeros at rest)
y = np.array([0.0, 0.0, 0.0, 0.0])

# Storage lists to record time history for plotting
force = []
y1 = []  # Car Body Displacement
y2 = []  # Car Body Velocity
y3 = []  # Wheel Displacement
y4 = []  # Wheel Velocity

# =============================================================================
# 5. SIMULATION LOOP (Transient "Haversine-like" Bump Profile)
# =============================================================================
ts = 5  # The exact second (5.0s) the vehicle encounters the bump

for t in T:
    # Phase 1: Flat surface before reaching the bump
    if 0 < t <= ts:
        F[3] = 0
        
    # Phase 2: Traveling over the smooth transient bump (lasts for 2 seconds)
    elif ts < t < (ts + 2):
        # A Cosine-based (raised cosine/haversine) shape function.
        # This provides a smooth ramp up and ramp down to prevent non-physical shocks.
        #F[3] = 0.5*(k2/m2)*a*(1-cos(w*(t-ts)))
        F[3] = 0.5 * F0 * (1 - np.cos(1 * (t - ts)))
        #F[3] = 0.5*F0*sin(t)
        
    # Phase 3: Flat surface after exiting the bump (free vibration/settling time)
    else:
        F[3] = 0
        
    # Forward Euler Numerical Integration Step:
    # Next State = Current State + (Rate of Change * Time Step)
    y = y + t_step * (A.dot(y) + F)
    
    # Store state histories
    y1.append(y[0])  # Car body displacement
    y2.append(y[1])
    y3.append(y[2])  # Wheel displacement
    y4.append(y[3])
    
    # Store forcing history
    force.append(F[3])

# =============================================================================
# 6. GRAPHICAL RENDER & PLOTTING
# =============================================================================
plt.figure(figsize=(10, 5))

# Plot the primary output: vehicle body displacement over time
plt.plot(T, y1, label='car displacement', linewidth=1, color='red')

# Clean, structured grid styling
plt.grid(linestyle='--', color='black')
plt.legend()

# Configure axis viewing limits
plt.xlim(T[0], T[-1])
plt.ylim(-a, a)      # Scales display window exactly to the road amplitude boundary

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from numpy import sin, cos, pi
from matplotlib.widgets import Slider, Button

# =============================================================================
# 1. SETUP PARAMETERS & STATE-SPACE SYSTEM
# =============================================================================
# Constant parameters
m2 = 30       # Unsprung mass (Wheel/Axle, kg)
k1 = 10000    # Suspension stiffness (N/m)
k2 = 40000    # Tire stiffness (N/m)
c2 = 0        # Tire damping (N-s/m)
a = 0.2       # Bump amplitude (m)
ts = 5        # Bump start time (s)

# Adjustable initial parameters for sliders
m1_init = 400    # Initial Sprung mass (Car body, kg)
c1_init = 1000   # Initial Suspension damping (N-s/m)

# Time array
t_step = 0.01
T = np.arange(0, 20 + t_step, t_step)

# Calculation function for the dynamic response loop
def simulate_quarter_car(m1, c1):
    # State space matrix A depends on the adjustable m1 and c1 variables
    A = np.array([[0.0, 1.0, 0.0, 0.0],
                  [-k1/m1, -c1/m1, k1/m1, c1/m1],
                  [0.0, 0.0, 0.0, 1.0],
                  [k1/m2, c1/m2, -(k1+k2)/m2, -(c1+c2)/m2]])
    
    F0 = k2 * a / m2
    F = np.array([0.0, 0.0, 0.0, 0.0])
    y = np.array([0.0, 0.0, 0.0, 0.0])
    
    y1_hist = []
    y3_hist = []
    
    for t in T:
        # Generate smooth bump event between ts and ts+2 seconds
        if ts < t < (ts + 2):
            F[3] = 0.5 * F0 * (1 - np.cos(1 * (t - ts)))
        else:
            F[3] = 0.0
            
        # Numerical Forward-Euler Step
        y = y + t_step * (A.dot(y) + F)
        y1_hist.append(y[0])  # Vehicle Displacement
        y3_hist.append(y[2])  # Wheel Displacement
        
    return y1_hist, y3_hist

# =============================================================================
# 2. RENDER THE INITIAL PLOT FRAME
# =============================================================================
fig, ax = plt.subplots(figsize=(10, 6))
plt.subplots_adjust(bottom=0.25)  # Leave room for sliders at the bottom

# Run initial simulation
y1_init, y3_init = simulate_quarter_car(m1_init, c1_init)

# Create plot line objects
line_car, = ax.plot(T, y1_init, label='Car Displacement ($y_1$)', linewidth=2, color='red')
line_wheel, = ax.plot(T, y3_init, label='Wheel Displacement ($y_3$)', linewidth=1.5, color='blue', alpha=0.7)

ax.grid(linestyle='--', color='black', alpha=0.3)
ax.legend(loc='upper right')
ax.set_xlim(T[0], T[-1])
ax.set_ylim(-a * 1.5, a * 1.5)
ax.set_xlabel('Time (s)')
ax.set_ylabel('Displacement (m)')
ax.set_title('Interactive Quarter-Car Transient Response')

# =============================================================================
# 3. INTERACTIVE SLIDER CONTROLS
# =============================================================================
# Define axes regions for sliders
ax_c1 = plt.axes([0.2, 0.12, 0.65, 0.03])
ax_m1 = plt.axes([0.2, 0.05, 0.65, 0.03])

# Create Sliders (Damping range: 100 to 5000 | Mass range: 150 to 1000)
slider_c1 = Slider(ax_c1, 'Damping ($c_1$)', 100.0, 5000.0, valinit=c1_init, valfmt='%0.0f N-s/m')
slider_m1 = Slider(ax_m1, 'Sprung Mass ($m_1$)', 150.0, 1000.0, valinit=m1_init, valfmt='%0.0f kg')

# Update event triggered whenever a slider is moved
def update(val):
    current_m1 = slider_m1.val
    current_c1 = slider_c1.val
    
    # Recalculate states
    y1_updated, y3_updated = simulate_quarter_car(current_m1, current_c1)
    
    # Update active lines
    line_car.set_ydata(y1_updated)
    line_wheel.set_ydata(y3_updated)
    
    fig.canvas.draw_idle()  # Redraw canvas cleanly

# Bind sliders to update function
slider_c1.on_changed(update)
slider_m1.on_changed(update)

plt.show()
```

```python
#AI

import numpy as np
import matplotlib.pyplot as plt
from numpy import sin, cos, pi
import tkinter as tk
from tkinter import ttk
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg

class QuarterCarApp:
    def __init__(self, root):
        self.root = root
        self.root.title("Quarter-Car Simulation Dashboard")
        self.root.geometry("1100x650")

        # --- Default Physical Parameters ---
        self.m1 = 400.0   # Sprung mass (kg)
        self.m2 = 30.0    # Unsprung mass (kg)
        self.k1 = 10e3    # Suspension stiffness (N/m)
        self.k2 = 40e3    # Tire stiffness (N/m)
        self.c1 = 1000.0  # Suspension damping (N-s/m)
        self.c2 = 0.0     # Tire damping (N-s/m)

        # --- Simulation Time Setup ---
        self.t_step = 0.01
        self.T = np.arange(0, 20, self.t_step)

        # --- Create GUI Layout ---
        self.create_widgets()
        self.run_simulation()

    def create_widgets(self):
        # Control Panel Frame
        control_frame = ttk.LabelFrame(self.root, text=" Control Parameters ", padding=15)
        control_frame.pack(side=tk.LEFT, fill=tk.Y, padx=15, pady=15)

        # 1. Road Profile Type Selection
        ttk.Label(control_frame, text="Road Profile Type:", font=('Arial', 10, 'bold')).pack(anchor=tk.W, pady=(0, 5))
        self.profile_type = tk.StringVar(value="Bump")
        profile_combo = ttk.Combobox(control_frame, textvariable=self.profile_type, values=["Bump", "Dip"], state="readonly")
        profile_combo.pack(fill=tk.X, pady=(0, 15))
        profile_combo.bind("<<ComboboxSelected>>", lambda e: self.run_simulation())

        # Helper function to create sliders
        def create_slider(parent, label, min_val, max_val, init_val, res):
            ttk.Label(parent, text=label).pack(anchor=tk.W)
            var = tk.DoubleVar(value=init_val)
            slider = ttk.Scale(parent, from_=min_val, to=max_val, variable=var, orient=tk.HORIZONTAL)
            slider.pack(fill=tk.X, pady=(0, 10))
            lbl_val = ttk.Label(parent, text=f"Value: {init_val:.2f}")
            lbl_val.pack(anchor=tk.E, pady=(0, 10))
            
            def update_val(*args):
                lbl_val.config(text=f"Value: {var.get():.2f}")
                self.run_simulation()
            
            var.trace_add("write", update_val)
            return var

        # 2. Sliders for Road Geometry & Speed
        self.var_a = create_slider(control_frame, "Amplitude / Depth (m)", 0.05, 0.5, 0.2, 0.01)
        self.var_v = create_slider(control_frame, "Forward Velocity (m/s)", 0.2, 5.0, 0.85, 0.05)
        self.var_d = create_slider(control_frame, "Profile Width (m)", 0.2, 3.0, 0.8, 0.1)
        self.var_ts = create_slider(control_frame, "Hit Time Start (s)", 0.5, 5.0, 2.5, 0.1)

        # Plot Window Frame
        self.plot_frame = ttk.Frame(self.root)
        self.plot_frame.pack(side=tk.RIGHT, fill=tk.BOTH, expand=True, padx=15, pady=15)

        # Initialize Matplotlib Figure
        self.fig, self.ax = plt.subplots(figsize=(8, 5), dpi=100)
        self.canvas = FigureCanvasTkAgg(self.fig, master=self.plot_frame)
        self.canvas.get_tk_widget().pack(fill=tk.BOTH, expand=True)

    def run_simulation(self):
        # Fetch current UI parameters
        a = self.var_a.get()
        v = self.var_v.get()
        d = self.var_d.get()
        ts = self.var_ts.get()
        
        # Adjust amplitude orientation based on selected type (Dip is a negative bump)
        if self.profile_type.get() == "Dip":
            a = -a

        # State-Space Formulation
        A_mat = np.array([
            [0.0, 1.0, 0.0, 0.0],
            [-self.k1/self.m1, -self.c1/self.m1, self.k1/self.m1, self.c1/self.m1],
            [0.0, 0.0, 0.0, 1.0],
            [self.k1/self.m2, self.c1/self.m2, -(self.k1+self.k2)/self.m2, -(self.c1+self.c2)/self.m2]
        ])

        F = np.array([0.0, 0.0, 0.0, 0.0])
        x = np.array([0.0, 0.0, 0.0, 0.0])

        w = 2 * pi * (v / d)

        d1, d2, road_prof = [], [], []

        # Simulation Loop (Forward Euler Integration)
        for t in self.T:
            if ts < t < (ts + d/v):
                z_r = 0.5 * a * (1 - cos(w * (t - ts)))
                F[3] = (self.k2 / self.m2) * z_r
            else:
                z_r = 0.0
                F[3] = 0.0

            x = x + self.t_step * (A_mat.dot(x) + F)

            d1.append(x[0])
            d2.append(x[2])
            road_prof.append(z_r)

        # Update Plots
        self.ax.clear()
        self.ax.plot(self.T, road_prof, label=f'Road Profile ({self.profile_type.get()})', color='gray', linestyle=':', linewidth=2)
        self.ax.plot(self.T, d2, label='Wheel Displacement', color='orange', alpha=0.8)
        self.ax.plot(self.T, d1, label='Vehicle Body Displacement', color='blue', linewidth=2)

        self.ax.grid(linestyle='--', color='k', alpha=0.3)
        self.ax.legend(loc='upper right')
        self.ax.set_xlim(self.T[0], self.T[-1])
        
        # Dynamic y-axis scaling based on profile choice
        if self.profile_type.get() == "Dip":
            self.ax.set_ylim(-self.var_a.get() * 1.5, self.var_a.get() * 1.5)
        else:
            self.ax.set_ylim(-self.var_a.get() * 0.5, self.var_a.get() * 2.5)

        self.ax.set_xlabel('Time (s)')
        self.ax.set_ylabel('Displacement / Height (m)')
        self.ax.set_title(f'Quarter-Car System Dynamics Contextualised to a Road {self.profile_type.get()}')
        
        self.canvas.draw()

if __name__ == "__main__":
    root = tk.Tk()
    app = QuarterCarApp(root)
    root.mainloop()
```
