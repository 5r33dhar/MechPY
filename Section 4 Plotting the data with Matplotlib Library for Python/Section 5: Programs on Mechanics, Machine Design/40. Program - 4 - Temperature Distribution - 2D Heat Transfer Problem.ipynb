```python
import numpy as np
import tkinter as tk
from tkinter import ttk
import matplotlib.pyplot as plt
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg

class HeatEquationApp:
    def __init__(self, root):
        self.root = root
        self.root.title("2D Heat Transfer Animation")
        
        # --- Physical System Configuration (From Your Given Parameters) ---
        self.xdim = 10         # Number of grid columns
        self.ydim = 10         # Number of grid rows
        self.niter = 100       # Total iteration limit for the solver
        self.current_iter = 0  # Counter tracking animation steps
        self.is_running = False # Control flag to play/pause the animation
        
        # Boundary Condition values
        self.t1 = 80           # Left boundary temperature
        self.t2 = 60           # Top boundary temperature
        self.t3 = 45           # Right boundary temperature
        self.t4 = 50           # Bottom boundary temperature
        self.t_guess = 25      # Initial interior point guess
        
        # --- Grid Initialization ---
        # Allocate a 2D temperature array and fill it with the initial guess
        self.T = np.full((self.ydim, self.xdim), float(self.t_guess))
        
        # Set persistent values on the outside edge boundaries
        self.T[0, :] = self.t2      # Top Edge
        self.T[-1, :] = self.t4     # Bottom Edge
        self.T[:, 0] = self.t1      # Left Edge
        self.T[:, -1] = self.t3     # Right Edge
        
        # Average out the four physical corners to handle step jumps smoothly
        self.T[0, 0] = (self.t2 + self.t1) / 2
        self.T[0, -1] = (self.t2 + self.t3) / 2
        self.T[-1, 0] = (self.t4 + self.t1) / 2
        self.T[-1, -1] = (self.t4 + self.t3) / 2
        
        # --- UI Layout Construction ---
        # Create a top header section for control widgets
        control_frame = ttk.Frame(root, padding=10)
        control_frame.pack(side=tk.TOP, fill=tk.X)
        
        self.start_btn = ttk.Button(control_frame, text="Start Animation", command=self.toggle_animation)
        self.start_btn.pack(side=tk.LEFT, padx=5)
        
        ttk.Button(control_frame, text="Reset Grid", command=self.reset_simulation).pack(side=tk.LEFT, padx=5)
        
        # Text label displaying current computational step to the user
        self.status_label = ttk.Label(control_frame, text="Iteration: 0 / 100 (Initialized)")
        self.status_label.pack(side=tk.RIGHT, padx=10)
        
        # --- Matplotlib Embedded Figure Configuration ---
        self.fig, self.ax = plt.subplots(figsize=(6, 5))
        
        # Draw the initial state contour map (80 levels provides smooth blending)
        self.contour = self.ax.contourf(self.T, 80, cmap='jet', vmin=min(self.t1, self.t2, self.t3, self.t4, self.t_guess), vmax=max(self.t1, self.t2, self.t3, self.t4))
        self.cbar = self.fig.colorbar(self.contour, ax=self.ax, label='Temperature (°C)')
        
        self.ax.set_title("Real-Time Temperature Dissemination")
        self.ax.set_xlabel("X Coordinate")
        self.ax.set_ylabel("Y Coordinate")
        self.ax.grid(color='k', linestyle='--', alpha=0.3)
        self.ax.invert_yaxis() # Match visual layout directly with matrix storage orientation
        
        # Embed the Matplotlib figure canvas frame directly inside the Tkinter loop
        self.canvas = FigureCanvasTkAgg(self.fig, master=root)
        self.canvas.get_tk_widget().pack(side=tk.BOTTOM, fill=tk.BOTH, expand=True)

    def update_heat_step(self):
        """Calculates one localized Jacobi explicit pass and updates the active graphic frame."""
        if not self.is_running or self.current_iter >= self.niter:
            if self.current_iter >= self.niter:
                self.is_running = False
                self.start_btn.config(text="Finished")
                self.status_label.config(text=f"Iteration: {self.current_iter}/{self.niter} (Steady-State Reached)")
            return
            
        # Create an immutable snapshot copy of the previous state array
        T_old = self.T.copy()
        
        # Apply the Finite Difference discrete update rule to all internal cells
        for i in range(1, self.ydim - 1):
            for j in range(1, self.xdim - 1):
                self.T[i, j] = 0.25 * (T_old[i+1, j] + T_old[i-1, j] + T_old[i, j+1] + T_old[i, j-1])
        
        self.current_iter += 1
        
        # Clear old contours to prevent memory leaking stacks inside the graph layout
        for coll in self.ax.collections:
            coll.remove()
            
        # Render the updated computational state frame
        self.contour = self.ax.contourf(self.T, 80, cmap='jet', vmin=25, vmax=80)
        self.status_label.config(text=f"Iteration: {self.current_iter} / {self.niter}")
        self.canvas.draw()
        
        # Queue up the next generation frame step loop pass in 50 milliseconds
        self.root.after(50, self.update_heat_step)

    def toggle_animation(self):
        """Switches playback processing flow states back and forth seamlessly."""
        if self.current_iter >= self.niter:
            return
        self.is_running = not self.is_running
        if self.is_running:
            self.start_btn.config(text="Pause")
            self.update_heat_step()
        else:
            self.start_btn.config(text="Resume")

    def reset_simulation(self):
        """Wipes the active grid and forces calculations back down to step zero baseline conditions."""
        self.is_running = False
        self.current_iter = 0
        self.start_btn.config(text="Start Animation")
        self.status_label.config(text="Iteration: 0 / 100 (Reset Complete)")
        
        # Reinitialize base matrix state arrays
        self.T.fill(self.t_guess)
        self.T[0, :] = self.t2
        self.T[-1, :] = self.t4
        self.T[:, 0] = self.t1
        self.T[:, -1] = self.t3
        self.T[0, 0] = (self.t2 + self.t1) / 2
        self.T[0, -1] = (self.t2 + self.t3) / 2
        self.st_0 = (self.t4 + self.t1) / 2
        self.T[-1, -1] = (self.t4 + self.t3) / 2
        
        # Force canvas graphic wipe pass update
        for coll in self.ax.collections:
            coll.remove()
        self.contour = self.ax.contourf(self.T, 80, cmap='jet', vmin=25, vmax=80)
        self.canvas.draw()

# --- Program Entry Point Execution Loop ---
if __name__ == "__main__":
    main_window = tk.Tk()
    app = HeatEquationApp(main_window)
    main_window.mainloop()
```

```python
import numpy as np
import tkinter as tk
from tkinter import ttk
import matplotlib.pyplot as plt
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg

class AdvancedHeatEquationApp:
    def __init__(self, root):
        self.root = root
        self.root.title("2D Heat Transfer Simulator with Live Analytics")
        
        # --- Physical System Configuration Variables ---
        self.xdim = 10         # Grid columns
        self.ydim = 10         # Grid rows
        self.current_iter = 0  # Frame counter
        self.is_running = False # Playback state tracker
        self.max_iter = 150    # Upper bound step threshold
        
        # Arrays to monitor convergence statistics
        self.iterations_history = []
        self.error_history = []
        
        # --- UI Window Structure ---
        # Sidebar for configuration inputs
        sidebar = ttk.Frame(root, padding=15, width=200)
        sidebar.pack(side=tk.LEFT, fill=tk.Y)
        
        # Main graphic layout window zone
        plot_frame = ttk.Frame(root, padding=10)
        plot_frame.pack(side=tk.RIGHT, fill=tk.BOTH, expand=True)
        
        # --- Sidebar Configurations Fields (Dynamic Control Variables) ---
        ttk.Label(sidebar, text="Grid Inputs (°C)", font=('Helvetica', 12, 'bold')).pack(anchor=tk.W, pady=(0,10))
        
        self.t1_var = tk.StringVar(value="80")
        self.t2_var = tk.StringVar(value="120")
        self.t3_var = tk.StringVar(value="60")
        self.t4_var = tk.StringVar(value="45")
        self.tg_var = tk.StringVar(value="25")
        
        self.create_input_field(sidebar, "Left Edge (T1):", self.t1_var)
        self.create_input_field(sidebar, "Top Edge (T2):", self.t2_var)
        self.create_input_field(sidebar, "Right Edge (T3):", self.t3_var)
        self.create_input_field(sidebar, "Bottom Edge (T4):", self.t4_var)
        self.create_input_field(sidebar, "Initial Guess:", self.tg_var)
        
        # Animation and Grid State controls
        ttk.Separator(sidebar, orient='horizontal').pack(fill='x', pady=15)
        
        self.start_btn = ttk.Button(sidebar, text="Start Animation", command=self.toggle_animation)
        self.start_btn.pack(fill='x', pady=5)
        
        ttk.Button(sidebar, text="Apply Changes & Reset", command=self.reset_simulation).pack(fill='x', pady=5)
        
        self.status_label = ttk.Label(sidebar, text="Iteration: 0\nStatus: Paused", justify=tk.LEFT)
        self.status_label.pack(anchor=tk.W, pady=15)
        
        # --- Matplotlib Multi-Panel Embedded Layout Configuration ---
        # Set up a dual-axis side-by-side plot layout frame
        self.fig, (self.ax_heat, self.ax_error) = plt.subplots(1, 2, figsize=(11, 5))
        self.fig.tight_layout(pad=4.0)
        
        # Initialize calculations and draw base canvases
        self.read_inputs_and_initialize_grid()
        self.init_plots()
        
        # Mount the Matplotlib component inside the Tkinter loop window
        self.canvas = FigureCanvasTkAgg(self.fig, master=plot_frame)
        self.canvas.get_tk_widget().pack(fill=tk.BOTH, expand=True)

    def create_input_field(self, parent, label_text, text_var):
        """Helper to quickly scaffold tidy labels and structural input text fields."""
        frame = ttk.Frame(parent)
        frame.pack(fill='x', pady=4)
        ttk.Label(frame, text=label_text, width=15, anchor='w').pack(side=tk.LEFT)
        ttk.Entry(frame, textvariable=text_var, width=8).pack(side=tk.RIGHT)

    def read_inputs_and_initialize_grid(self):
        """Extracts text inputs cleanly and sets up initial grid states safely."""
        try:
            t1 = float(self.t1_var.get())
            t2 = float(self.t2_var.get())
            t3 = float(self.t3_var.get())
            t4 = float(self.t4_var.get())
            tg = float(self.tg_var.get())
        except ValueError:
            # Fallbacks to default boundaries if string conversions break
            t1, t2, t3, t4, tg = 80.0, 60.0, 45.0, 50.0, 25.0
            
        self.T = np.full((self.ydim, self.xdim), tg)
        self.T[0, :] = t2
        self.T[-1, :] = t4
        self.T[:, 0] = t1
        self.T[:, -1] = t3
        
        # Clean corner interpolation updates
        self.T[0, 0] = (t2 + t1) / 2
        self.T[0, -1] = (t2 + t3) / 2
        self.T[-1, 0] = (t4 + t1) / 2
        self.T[-1, -1] = (t4 + t3) / 2

    def init_plots(self):
        """Builds static styling foundations for both visualization charts."""
        # 1. Heatmap configuration
        self.ax_heat.clear()
        self.contour = self.ax_heat.contourf(self.T, 80, cmap='jet', vmin=0, vmax=100)
        self.ax_heat.set_title("Thermal Heatmap Profile")
        self.ax_heat.invert_yaxis()
        self.ax_heat.grid(True, linestyle='--', alpha=0.3)
        
        if not hasattr(self, 'cbar'):
            self.cbar = self.fig.colorbar(self.contour, ax=self.ax_heat, label='Temp (°C)')
            
        # 2. Convergence/Error tracking line graph configuration
        self.ax_error.clear()
        self.error_line, = self.ax_error.plot([], [], 'r-', linewidth=2, label='Mean System Shift')
        self.ax_error.set_title("Energy Error Convergence Rate")
        self.ax_error.set_xlabel("Iterations Steps")
        self.ax_error.set_ylabel("Absolute Maximum ΔT")
        self.ax_error.set_yscale('log') # Logarithmic representation profiles decay changes beautifully
        self.ax_error.grid(True, which="both", linestyle=':', alpha=0.5)
        self.ax_error.legend()

    def update_simulation_step(self):
        """Executes one math frame evaluation sweep, calculating absolute difference shifts."""
        if not self.is_running or self.current_iter >= self.max_iter:
            if self.current_iter >= self.max_iter:
                self.is_running = False
                self.start_btn.config(text="Finished")
                self.status_label.config(text=f"Iteration: {self.current_iter}\nStatus: Converged")
            return

        T_old = self.T.copy()
        
        # Calculate localized system states over inner index fields
        for i in range(1, self.ydim - 1):
            for j in range(1, self.xdim - 1):
                self.T[i, j] = 0.25 * (T_old[i+1, j] + T_old[i-1, j] + T_old[i, j+1] + T_old[i, j-1])
                
        # Metric tracking: Find peak absolute difference across entire mesh matrix
        max_delta = np.max(np.abs(self.T - T_old))
        
        self.current_iter += 1
        self.iterations_history.append(self.current_iter)
        self.error_history.append(max_delta)
        
        # Refresh heatmap plot profiles
        for coll in self.ax_heat.collections:
            coll.remove()
        self.contour = self.ax_heat.contourf(self.T, 80, cmap='jet', vmin=0, vmax=100)
        
        # Dynamic line extension adjustments
        self.error_line.set_data(self.iterations_history, self.error_history)
        self.ax_error.set_xlim(0, max(10, self.current_iter + 5))
        self.ax_error.set_ylim(min(self.error_history)*0.5, max(self.error_history)*1.5)
        
        # Update text labels
        self.status_label.config(text=f"Iteration: {self.current_iter}\nMax ΔT: {max_delta:.5f}\nStatus: Processing")
        self.canvas.draw()
        
        # Reschedule next execution cycle pass
        self.root.after(40, self.update_simulation_step)

    def toggle_animation(self):
        """Handles seamless run state toggles."""
        if self.current_iter >= self.max_iter:
            return
        self.is_running = not self.is_running
        if self.is_running:
            self.start_btn.config(text="Pause")
            self.update_simulation_step()
        else:
            self.start_btn.config(text="Resume")
            self.status_label.config(text=f"Iteration: {self.current_iter}\nStatus: Paused")

    def reset_simulation(self):
        """Flushes storage tracking pipelines completely and resets using newly set custom inputs."""
        self.is_running = False
        self.current_iter = 0
        self.iterations_history.clear()
        self.error_history.clear()
        self.start_btn.config(text="Start Animation")
        self.status_label.config(text="Iteration: 0\nStatus: Grid Reset")
        
        self.read_inputs_and_initialize_grid()
        self.init_plots()
        self.canvas.draw()

if __name__ == "__main__":
    window = tk.Tk()
    app = AdvancedHeatEquationApp(window)
    window.mainloop()
```

```python
import numpy as np
import tkinter as tk
from tkinter import ttk
import matplotlib.pyplot as plt
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg

class RealTimeDynamicHeatApp:
    def __init__(self, root):
        self.root = root
        self.root.title("2D Heat Transfer - Live Dynamic Input Controls")
        
        # --- Physical System Configuration Variables ---
        # Grid sizes scaled up safely thanks to new vectorized NumPy calculation speeds
        self.xdim = 30         
        self.ydim = 30         
        self.current_iter = 0  
        self.is_running = False 
        self.max_iter = 500    # Higher iteration bound for high-res grids
        
        # Performance/Analytics tracking stores
        self.iterations_history = []
        self.error_history = []
        
        # --- UI Window Grid Setup ---
        sidebar = ttk.Frame(root, padding=15, width=220)
        sidebar.pack(side=tk.LEFT, fill=tk.Y)
        
        plot_frame = ttk.Frame(root, padding=10)
        plot_frame.pack(side=tk.RIGHT, fill=tk.BOTH, expand=True)
        
        # --- Real-Time Boundary Variable Triggers ---
        ttk.Label(sidebar, text="Live Edge Tuners (°C)", font=('Helvetica', 11, 'bold')).pack(anchor=tk.W, pady=(0,10))
        
        # Tkinter StringVars to store user inputs
        self.t1_var = tk.StringVar(value="80")
        self.t2_var = tk.StringVar(value="120")
        self.t3_var = tk.StringVar(value="60")
        self.t4_var = tk.StringVar(value="50")
        self.tg_var = tk.StringVar(value="45")
        
        # Attach observers ('write' mode) that run recalculations instantly as you type
        for var in [self.t1_var, self.t2_var, self.t3_var, self.t4_var, self.tg_var]:
            var.trace_add("write", lambda name, index, mode: self.live_parameter_update())
            
        # Draw control boxes inside sidebar layout zone
        self.create_spin_control(sidebar, "Left Bound (T1):", self.t1_var)
        self.create_spin_control(sidebar, "Top Bound (T2):", self.t2_var)
        self.create_spin_control(sidebar, "Right Bound (T3):", self.t3_var)
        self.create_spin_control(sidebar, "Bottom Bound (T4):", self.t4_var)
        self.create_spin_control(sidebar, "Core Guess (Tg):", self.tg_var)
        
        # Playback Management Controls
        ttk.Separator(sidebar, orient='horizontal').pack(fill='x', pady=15)
        
        self.start_btn = ttk.Button(sidebar, text="Play Simulation", command=self.toggle_animation)
        self.start_btn.pack(fill='x', pady=5)
        
        ttk.Button(sidebar, text="Clear & Restart", command=self.reset_simulation).pack(fill='x', pady=5)
        
        self.status_label = ttk.Label(sidebar, text="Iteration: 0\nStatus: Ready", justify=tk.LEFT)
        self.status_label.pack(anchor=tk.W, pady=15)
        
        # --- Matplotlib Canvas Initialization ---
        self.fig, (self.ax_heat, self.ax_error) = plt.subplots(1, 2, figsize=(11, 5))
        self.fig.tight_layout(pad=4.0)
        
        # Initial compute sweeps
        self.read_inputs_and_initialize_grid()
        self.init_plots()
        
        self.canvas = FigureCanvasTkAgg(self.fig, master=plot_frame)
        self.canvas.get_tk_widget().pack(fill=tk.BOTH, expand=True)

    def create_spin_control(self, parent, label_text, text_var):
        """Builds a combination slider entry that allows instant live adjustments."""
        frame = ttk.Frame(parent)
        frame.pack(fill='x', pady=6)
        ttk.Label(frame, text=label_text, width=15, anchor='w').pack(side=tk.LEFT)
        # Using Spinbox gives an intuitive up/down arrow adjustment path
        spin = tk.Spinbox(frame, from_=0, to=120, increment=5, textvariable=text_var, width=6)
        spin.pack(side=tk.RIGHT)

    def read_inputs_and_initialize_grid(self):
        """Reads boundaries dynamically while filtering out incomplete string entries cleanly."""
        try:
            t1 = float(self.t1_var.get() or 0)
            t2 = float(self.t2_var.get() or 0)
            t3 = float(self.t3_var.get() or 0)
            t4 = float(self.t4_var.get() or 0)
            tg = float(self.tg_var.get() or 0)
        except ValueError:
            return # Ignore half-typed input strings until conversion checks out safely
            
        self.T = np.full((self.ydim, self.xdim), tg)
        self.T[0, :] = t2
        self.T[-1, :] = t4
        self.T[:, 0] = t1
        self.T[:, -1] = t3
        
        # Smooth corner node transitions
        self.T[0, 0] = (t2 + t1) / 2
        self.T[0, -1] = (t2 + t3) / 2
        self.T[-1, 0] = (t4 + t1) / 2
        self.T[-1, -1] = (t4 + t3) / 2

    def init_plots(self):
        """Plots initialization values structure frameworks layout zones."""
        self.ax_heat.clear()
        self.contour = self.ax_heat.contourf(self.T, 80, cmap='jet', vmin=0, vmax=120)
        self.ax_heat.set_title(f"Live Heatmap ({self.xdim}x{self.ydim} Grid)")
        self.ax_heat.invert_yaxis()
        self.ax_heat.grid(True, linestyle='--', alpha=0.2)
        
        if not hasattr(self, 'cbar'):
            self.cbar = self.fig.colorbar(self.contour, ax=self.ax_heat, label='Temp (°C)')
            
        self.ax_error.clear()
        self.error_line, = self.ax_error.plot([], [], 'g-', linewidth=2, label='Convergence Shift')
        self.ax_error.set_title("Energy Error Divergence Loss")
        self.ax_error.set_xlabel("Steps")
        self.ax_error.set_ylabel("Max Absolute ΔT")
        self.ax_error.set_yscale('log')
        self.ax_error.grid(True, which="both", linestyle=':', alpha=0.5)
        self.ax_error.legend()

    def live_parameter_update(self):
        """Pushes boundary modifications immediately down to array structures without stopping calculations."""
        try:
            t1 = float(self.t1_var.get() or 0)
            t2 = float(self.t2_var.get() or 0)
            t3 = float(self.t3_var.get() or 0)
            t4 = float(self.t4_var.get() or 0)
        except ValueError:
            return
            
        # Dynamically inject boundary values directly over active grid limits
        self.T[0, :] = t2
        self.T[-1, :] = t4
        self.T[:, 0] = t1
        self.T[:, -1] = t3
        self.T[0, 0] = (t2 + t1) / 2
        self.T[0, -1] = (t2 + t3) / 2
        self.T[-1, 0] = (t4 + t1) / 2
        self.T[-1, -1] = (t4 + t3) / 2

        # If simulation is paused, update the UI layout immediately to reflect structural adjustments
        if not self.is_running:
            self.refresh_graphics_view()

    def update_simulation_step(self):
        """Executes a highly optimized Vectorized NumPy slice matrix update loop sweep step."""
        if not self.is_running or self.current_iter >= self.max_iter:
            if self.current_iter >= self.max_iter:
                self.is_running = False
                self.start_btn.config(text="Finished")
                self.status_label.config(text=f"Iteration: {self.current_iter}\nStatus: Maximized Run")
            return

        T_old = self.T.copy()
        
        # --- Vectorized NumPy Stencil Operation ---
        # Replaces double-nested loops. Calculates the entire grid simultaneously via matrix shifts.
        self.T[1:-1, 1:-1] = 0.25 * (T_old[:-2, 1:-1] +  # Shift down (Top neighbors)
                                     T_old[2:, 1:-1] +   # Shift up (Bottom neighbors)
                                     T_old[1:-1, :-2] +  # Shift right (Left neighbors)
                                     T_old[1:-1, 2:])    # Shift left (Right neighbors)
                
        # Monitor energy variation step profiles
        max_delta = np.max(np.abs(self.T - T_old))
        
        self.current_iter += 1
        self.iterations_history.append(self.current_iter)
        self.error_history.append(max_delta if max_delta > 0 else 1e-10) # Prevent log zero clipping boundaries
        
        self.refresh_graphics_view()
        
        # Dynamic error boundary stretching line limits adjustments
        self.error_line.set_data(self.iterations_history, self.error_history)
        self.ax_error.set_xlim(0, max(10, self.current_iter + 5))
        self.ax_error.set_ylim(min(self.error_history)*0.5, max(self.error_history)*1.5)
        
        self.status_label.config(text=f"Iteration: {self.current_iter}\nMax ΔT: {max_delta:.5f}\nStatus: Running")
        
        # Queue next frame faster (25ms execution cadence)
        self.root.after(25, self.update_simulation_step)

    def refresh_graphics_view(self):
        """Wipes layer collections smoothly and updates the plot view frame canvas."""
        for coll in self.ax_heat.collections:
            coll.remove()
        self.contour = self.ax_heat.contourf(self.T, 80, cmap='jet', vmin=0, vmax=120)
        self.canvas.draw()

    def toggle_animation(self):
        """Handles seamless run state toggles."""
        if self.current_iter >= self.max_iter:
            return
        self.is_running = not self.is_running
        if self.is_running:
            self.start_btn.config(text="Pause")
            self.update_simulation_step()
        else:
            self.start_btn.config(text="Play Simulation")
            self.status_label.config(text=f"Iteration: {self.current_iter}\nStatus: Paused")

    def reset_simulation(self):
        """Wipes tracking metrics and resets grid configurations back to starting base profiles."""
        self.is_running = False
        self.current_iter = 0
        self.iterations_history.clear()
        self.error_history.clear()
        self.start_btn.config(text="Play Simulation")
        self.status_label.config(text="Iteration: 0\nStatus: Clean Slate")
        
        self.read_inputs_and_initialize_grid()
        self.init_plots()
        self.canvas.draw()

if __name__ == "__main__":
    window = tk.Tk()
    app = RealTimeDynamicHeatApp(window)
    window.mainloop()
```

```python
import numpy as np
import tkinter as tk
from tkinter import ttk

import matplotlib.pyplot as plt
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg


class RealTimeDynamicHeatApp:
    def __init__(self, root):
        self.root = root
        self.root.title("2D Heat Transfer - Live Dynamic Input Controls")
        self.root.geometry("1400x700")

        # -----------------------------
        # Simulation Parameters
        # -----------------------------
        self.xdim = 30
        self.ydim = 30

        self.current_iter = 0
        self.max_iter = 500
        self.is_running = False

        self.iterations_history = []
        self.error_history = []

        # -----------------------------
        # Layout
        # -----------------------------
        sidebar = ttk.Frame(root, padding=15, width=300)
        sidebar.pack(side=tk.LEFT, fill=tk.Y)

        plot_frame = ttk.Frame(root, padding=10)
        plot_frame.pack(side=tk.RIGHT, fill=tk.BOTH, expand=True)

        ttk.Label(
            sidebar,
            text="Temperature Controls (°C)",
            font=("Helvetica", 12, "bold")
        ).pack(anchor=tk.W, pady=(0, 10))

        # -----------------------------
        # Slider Variables
        # -----------------------------
        self.t1_var = tk.DoubleVar(value=80.0)
        self.t2_var = tk.DoubleVar(value=120.0)
        self.t3_var = tk.DoubleVar(value=60.0)
        self.t4_var = tk.DoubleVar(value=50.0)
        self.tg_var = tk.DoubleVar(value=45.0)

        for var in (
            self.t1_var,
            self.t2_var,
            self.t3_var,
            self.t4_var,
            self.tg_var,
        ):
            var.trace_add("write", lambda *args: self.live_parameter_update())

        # -----------------------------
        # Create Sliders
        # -----------------------------
        self.create_slider_control(
            sidebar,
            "Left Boundary (T1)",
            self.t1_var
        )

        self.create_slider_control(
            sidebar,
            "Top Boundary (T2)",
            self.t2_var
        )

        self.create_slider_control(
            sidebar,
            "Right Boundary (T3)",
            self.t3_var
        )

        self.create_slider_control(
            sidebar,
            "Bottom Boundary (T4)",
            self.t4_var
        )

        self.create_slider_control(
            sidebar,
            "Initial Guess (Tg)",
            self.tg_var
        )

        # Live values display
        self.temp_display = ttk.Label(
            sidebar,
            font=("Calibri", 10, "bold"),
            justify=tk.LEFT
        )
        self.temp_display.pack(pady=10)

        ttk.Separator(sidebar, orient="horizontal").pack(
            fill="x", pady=10
        )

        # -----------------------------
        # Buttons
        # -----------------------------
        self.start_btn = ttk.Button(
            sidebar,
            text="Play Simulation",
            command=self.toggle_animation
        )
        self.start_btn.pack(fill="x", pady=5)

        ttk.Button(
            sidebar,
            text="Reset Simulation",
            command=self.reset_simulation
        ).pack(fill="x", pady=5)

        self.status_label = ttk.Label(
            sidebar,
            text="Iteration: 0\nStatus: Ready",
            justify=tk.LEFT
        )
        self.status_label.pack(anchor=tk.W, pady=15)

        # -----------------------------
        # Matplotlib Figure
        # -----------------------------
        self.fig, (self.ax_heat, self.ax_error) = plt.subplots(
            1, 2, figsize=(11, 5)
        )

        self.fig.tight_layout(pad=4.0)

        self.read_inputs_and_initialize_grid()
        self.init_plots()

        self.canvas = FigureCanvasTkAgg(
            self.fig,
            master=plot_frame
        )

        self.canvas.get_tk_widget().pack(
            fill=tk.BOTH,
            expand=True
        )

        self.update_temperature_display()

    # --------------------------------------------------
    # Slider Widget
    # --------------------------------------------------
    def create_slider_control(self, parent, label_text, variable):
        frame = ttk.Frame(parent)
        frame.pack(fill="x", pady=8)

        header = ttk.Frame(frame)
        header.pack(fill="x")

        ttk.Label(
            header,
            text=label_text
        ).pack(side=tk.LEFT)

        value_label = ttk.Label(
            header,
            text=f"{variable.get():.1f} °C",
            font=("Courier", 10, "bold")
        )
        value_label.pack(side=tk.RIGHT)

        def update_label(value):
            value_label.config(
                text=f"{float(value):.1f} °C"
            )

        slider = ttk.Scale(
            frame,
            from_=0,
            to=120,
            orient=tk.HORIZONTAL,
            variable=variable,
            command=update_label
        )

        slider.pack(fill="x")

    # --------------------------------------------------
    # Grid Initialization
    # --------------------------------------------------
    def read_inputs_and_initialize_grid(self):

        t1 = self.t1_var.get()
        t2 = self.t2_var.get()
        t3 = self.t3_var.get()
        t4 = self.t4_var.get()
        tg = self.tg_var.get()

        self.T = np.full(
            (self.ydim, self.xdim),
            tg
        )

        self.T[0, :] = t2
        self.T[-1, :] = t4
        self.T[:, 0] = t1
        self.T[:, -1] = t3

        self.T[0, 0] = (t1 + t2) / 2
        self.T[0, -1] = (t2 + t3) / 2
        self.T[-1, 0] = (t1 + t4) / 2
        self.T[-1, -1] = (t3 + t4) / 2

    # --------------------------------------------------
    # Plot Initialization
    # --------------------------------------------------
    def init_plots(self):

        self.ax_heat.clear()

        self.contour = self.ax_heat.contourf(
            self.T,
            levels=80,
            cmap="jet",
            vmin=0,
            vmax=120
        )

        self.ax_heat.set_title(
            f"Heat Distribution ({self.xdim} x {self.ydim})"
        )

        self.ax_heat.invert_yaxis()

        if not hasattr(self, "cbar"):
            self.cbar = self.fig.colorbar(
                self.contour,
                ax=self.ax_heat,
                label="Temperature (°C)"
            )

        self.ax_error.clear()

        self.error_line, = self.ax_error.plot(
            [],
            [],
            "g-",
            linewidth=2,
            label="Convergence Error"
        )

        self.ax_error.set_title(
            "Convergence History"
        )

        self.ax_error.set_xlabel(
            "Iteration"
        )

        self.ax_error.set_ylabel(
            "Max ΔT"
        )

        self.ax_error.set_yscale(
            "log"
        )

        self.ax_error.grid(
            True,
            which="both",
            linestyle=":"
        )

        self.ax_error.legend()

    # --------------------------------------------------
    # Update Text Display
    # --------------------------------------------------
    def update_temperature_display(self):

        self.temp_display.config(
            text=(
                f"T1 = {self.t1_var.get():.1f} °C\n"
                f"T2 = {self.t2_var.get():.1f} °C\n"
                f"T3 = {self.t3_var.get():.1f} °C\n"
                f"T4 = {self.t4_var.get():.1f} °C\n"
                f"Tg = {self.tg_var.get():.1f} °C"
            )
        )

    # --------------------------------------------------
    # Live Slider Update
    # --------------------------------------------------
    def live_parameter_update(self):

        if not hasattr(self, "T"):
            return

        t1 = self.t1_var.get()
        t2 = self.t2_var.get()
        t3 = self.t3_var.get()
        t4 = self.t4_var.get()

        self.update_temperature_display()

        self.T[0, :] = t2
        self.T[-1, :] = t4
        self.T[:, 0] = t1
        self.T[:, -1] = t3

        self.T[0, 0] = (t1 + t2) / 2
        self.T[0, -1] = (t2 + t3) / 2
        self.T[-1, 0] = (t1 + t4) / 2
        self.T[-1, -1] = (t3 + t4) / 2

        if not self.is_running:
            self.refresh_graphics_view()

    # --------------------------------------------------
    # Simulation Step
    # --------------------------------------------------
    def update_simulation_step(self):

        if not self.is_running or self.current_iter >= self.max_iter:

            if self.current_iter >= self.max_iter:
                self.is_running = False

                self.start_btn.config(
                    text="Finished"
                )

            return

        T_old = self.T.copy()

        self.T[1:-1, 1:-1] = 0.25 * (
            T_old[:-2, 1:-1] +
            T_old[2:, 1:-1] +
            T_old[1:-1, :-2] +
            T_old[1:-1, 2:]
        )

        max_delta = np.max(
            np.abs(self.T - T_old)
        )

        self.current_iter += 1

        self.iterations_history.append(
            self.current_iter
        )

        self.error_history.append(
            max(max_delta, 1e-10)
        )

        self.error_line.set_data(
            self.iterations_history,
            self.error_history
        )

        self.ax_error.set_xlim(
            0,
            max(10, self.current_iter + 5)
        )

        self.ax_error.set_ylim(
            min(self.error_history) * 0.5,
            max(self.error_history) * 1.5
        )

        self.status_label.config(
            text=(
                f"Iteration: {self.current_iter}\n"
                f"Max ΔT: {max_delta:.6f}\n"
                f"Status: Running"
            )
        )

        self.refresh_graphics_view()

        self.root.after(
            25,
            self.update_simulation_step
        )

    # --------------------------------------------------
    # Refresh Plot
    # --------------------------------------------------
    def refresh_graphics_view(self):

        self.ax_heat.clear()

        self.contour = self.ax_heat.contourf(
            self.T,
            levels=80,
            cmap="jet",
            vmin=0,
            vmax=120
        )

        self.ax_heat.set_title(
            f"Heat Distribution ({self.xdim} x {self.ydim})"
        )

        self.ax_heat.invert_yaxis()

        self.canvas.draw_idle()

    # --------------------------------------------------
    # Start / Pause
    # --------------------------------------------------
    def toggle_animation(self):

        if self.current_iter >= self.max_iter:
            return

        self.is_running = not self.is_running

        if self.is_running:
            self.start_btn.config(
                text="Pause"
            )
            self.update_simulation_step()

        else:
            self.start_btn.config(
                text="Play Simulation"
            )

            self.status_label.config(
                text=f"Iteration: {self.current_iter}\nStatus: Paused"
            )

    # --------------------------------------------------
    # Reset
    # --------------------------------------------------
    def reset_simulation(self):

        self.is_running = False
        self.current_iter = 0

        self.iterations_history.clear()
        self.error_history.clear()

        self.start_btn.config(
            text="Play Simulation"
        )

        self.read_inputs_and_initialize_grid()
        self.init_plots()
        self.refresh_graphics_view()

        self.status_label.config(
            text="Iteration: 0\nStatus: Reset"
        )


if __name__ == "__main__":
    window = tk.Tk()
    app = RealTimeDynamicHeatApp(window)
    window.mainloop()
```
