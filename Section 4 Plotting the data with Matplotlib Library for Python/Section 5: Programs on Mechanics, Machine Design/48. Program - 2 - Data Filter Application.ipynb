```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

bearing_data = pd.read_csv("od.csv",skiprows=6)
bearing_data.head(10)
```

```python
t = bearing_data["Time"]
accl = bearing_data["Acc"]

plt.plot(t,accl)
plt.show()
```

```python
from scipy import signal

my_filt = signal.butter(5, 25, 'lp', fs=750,output='sos')
filt_accl = signal.sosfilt(my_filt, accl)

plt.figure(figsize=(10,7.5))

plt.plot(t,accl,label="Orignal Acceleation",alpha=0.5)
plt.plot(t,filt_accl, label="Filtered Acceleration")

plt.legend(fontsize = 15)
plt.xlim(0,0.05)

plt.show()
```
