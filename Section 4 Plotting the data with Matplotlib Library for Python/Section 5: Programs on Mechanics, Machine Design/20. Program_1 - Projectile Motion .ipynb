## Telemetry Output

```python
import numpy as np
import matplotlib.pyplot as plt
from numpy import sin, cos, pi

def get_projectile_trajectory(u0, theta, g=9.81, num_points=100):
    """
    Calculates coordinates and telemetry for a projectile trajectory.
    Returns: sx, sy (arrays) and telemetry (dictionary)
    """
    theta_rad = theta * pi / 180
    
    ux = u0 * cos(theta_rad)
    uy = u0 * sin(theta_rad)
    
    # Kinematic Equations for exact telemetry values
    t_total = 2 * uy / g
    max_range = ux * t_total
    peak_height = (uy ** 2) / (2 * g)
    
    # Generate spatial arrays
    t = np.linspace(0, t_total, num_points)
    sx = ux * t
    sy = uy * t - 0.5 * g * t**2
    
    # Package additional metrics
    telemetry = {
        "flight_time_s": t_total,
        "max_range_m": max_range,
        "peak_height_m": peak_height
    }
    
    return sx, sy, telemetry

# Constants
u0 = 40

# Generate trajectories and extract telemetry
sx1, sy1, tel1 = get_projectile_trajectory(u0, theta=45)
sx2, sy2, tel2 = get_projectile_trajectory(u0, theta=30)

# Print Telemetry Report to Console
print("--- TELEMETRY REPORT ---")
print(f"Launch Angle: 45° | Max Range: {tel1['max_range_m']:.2f}m | Peak Height: {tel1['peak_height_m']:.2f}m | Flight Time: {tel1['flight_time_s']:.2f}s")
print(f"Launch Angle: 30° | Max Range: {tel2['max_range_m']:.2f}m | Peak Height: {tel2['peak_height_m']:.2f}m | Flight Time: {tel2['flight_time_s']:.2f}s")

# Plotting configuration
plt.figure(figsize=(8, 5))
plt.plot(sx1, sy1, label='theta = 45°')
plt.plot(sx2, sy2, label='theta = 30°')

# Visually anchor peak heights on the plot
plt.scatter(tel1['max_range_m']/2, tel1['peak_height_m'], color='blue', zorder=5)
plt.scatter(tel2['max_range_m']/2, tel2['peak_height_m'], color='orange', zorder=5)

plt.xlabel("Horizontal distance (m)")
plt.ylabel("Height (m)")
plt.ylim(0, 50)
plt.xlim(0, 175)
plt.grid(True, linestyle='--', alpha=0.6)
plt.title("Projectile Trajectories with Telemetry Tracking")
plt.legend()
plt.show()
```

## Add an interactive slider animation to see how changing angles dynamically alters the path

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.widgets import Slider
from numpy import sin, cos, pi



# Initial Configuration
u0 = 40
g = 9.81
initial_theta = 45

def get_projectile_trajectory(u0, theta, g=9.81, num_points=200):
    theta_rad = theta * pi / 180
    ux = u0 * cos(theta_rad)
    uy = u0 * sin(theta_rad)
    
    t_total = 2 * uy / g
    max_range = ux * t_total
    peak_height = (uy ** 2) / (2 * g)
    
    t = np.linspace(0, t_total, num_points)
    sx = ux * t
    sy = uy * t - 0.5 * g * t**2
    
    telemetry = {
        "flight_time_s": t_total,
        "max_range_m": max_range,
        "peak_height_m": peak_height
    }
    return sx, sy, telemetry

# Create Figure Window
fig, ax = plt.subplots(figsize=(8, 5.5))
plt.subplots_adjust(bottom=0.2) # Reserve space at bottom for the slider

# Generate baseline curve
sx, sy, tel = get_projectile_trajectory(u0, initial_theta)
line, = ax.plot(sx, sy, lw=2, color='royalblue', label=f'Trajectory')
peak_dot, = ax.plot(tel['max_range_m']/2, tel['peak_height_m'], 'ro', label='Apex Peak')

# Graph decoration
ax.set_xlabel("Horizontal distance (m)")
ax.set_ylabel("Height (m)")
ax.set_xlim(0, 175)
ax.set_ylim(0, 85) # Taller window to accommodate high 90-degree launches
ax.grid(True, linestyle='--', alpha=0.6)
ax.legend(loc='upper right')

# Build Live Telemetry Text Template
text_box = ax.text(5, 75, '', fontsize=10, bbox=dict(facecolor='white', alpha=0.8, boxstyle='round,pad=0.5'))

def update_text(tel, theta):
    text_box.set_text(
        f"Angle: {theta}°\n"
        f"Max Range: {tel['max_range_m']:.2f} m\n"
        f"Peak Height: {tel['peak_height_m']:.2f} m\n"
        f"Flight Time: {tel['flight_time_s']:.2f} s"
    )

update_text(tel, initial_theta)

# Add Slider Widget Object
slider_ax = plt.axes([0.15, 0.05, 0.7, 0.03]) # [left, bottom, width, height]
angle_slider = Slider(
    ax=slider_ax,
    label='Launch Angle (θ) ',
    valmin=0.0,
    valmax=90.0,
    valinit=initial_theta,
    valfmt='%1.1f°',
    color='royalblue'
)

# Callback listener to trigger recalculation on slider movement
def update(val):
    current_theta = angle_slider.val
    new_sx, new_sy, new_tel = get_projectile_trajectory(u0, current_theta)
    
    # Update visual markers
    line.set_xdata(new_sx)
    line.set_ydata(new_sy)
    peak_dot.set_data([new_tel['max_range_m']/2], [new_tel['peak_height_m']])
    
    # Refresh live text readouts
    update_text(new_tel, round(current_theta, 1))
    fig.canvas.draw_idle()

angle_slider.on_changed(update)
plt.show()
```
