```python
#given 
import numpy as np
import matplotlib.pyplot as plt

class DieselCycleSimulator:
    def __init__(self, bore, stroke, connecting_rod_length, compression_ratio):
        """
        Initialize diesel engine parameters
        
        Parameters:
        bore: cylinder bore diameter (m)
        stroke: piston stroke (m) 
        connecting_rod_length: connecting rod length (m)
        compression_ratio: compression ratio
        """
        self.bore = bore
        self.stroke = stroke
        self.connecting_rod_length = connecting_rod_length
        self.compression_ratio = compression_ratio
        
        # Calculate basic volumes
        self.swept_volume = (np.pi/4) * bore**2 * stroke
        self.clearance_volume = self.swept_volume / (compression_ratio - 1)
        self.total_volume = self.swept_volume + self.clearance_volume
        
    def instantaneous_volume(self, crank_angle):
        """
        Calculate instantaneous cylinder volume based on crank angle
        
        Parameters:
        crank_angle: crank angle in radians (0 = TDC)
        
        Returns:
        volume: instantaneous cylinder volume (m³)
        """
        # Corrected volume calculation
        R = self.stroke / 2  # crank radius
        L = self.connecting_rod_length
        
        # Distance from crank center to piston pin
        x = R * np.cos(crank_angle) + np.sqrt(L**2 - (R * np.sin(crank_angle))**2)
        
        # Distance from TDC to piston
        h = L + R - x
        
        # Instantaneous volume
        volume = self.clearance_volume + (np.pi/4) * self.bore**2 * h
        return volume
    
    def ideal_diesel_cycle(self, p1, T1, T3, gamma=1.4):
        """
        Calculate ideal diesel cycle thermodynamic states
        
        Parameters:
        p1: initial pressure (kPa)
        T1: initial temperature (K)
        T3: peak combustion temperature (K)
        gamma: specific heat ratio
        
        Returns:
        dict with cycle state points
        """
        # State 1: Start of compression (BDC)
        v1 = self.total_volume
        
        # State 2: End of compression (TDC)
        v2 = self.clearance_volume
        p2 = p1 * (self.compression_ratio)**gamma
        T2 = T1 * (self.compression_ratio)**(gamma-1)
        
        # State 3: End of heat addition (constant pressure)
        p3 = p2  # constant pressure heat addition
        v3 = v2 * T3/T2  # from ideal gas law at constant pressure
        
        # State 4: End of expansion (BDC)
        v4 = v1
        p4 = p3 * (v3/v4)**gamma
        T4 = T3 * (v3/v4)**(gamma-1)
        
        return {
            'states': {
                1: {'p': p1, 'v': v1, 'T': T1},
                2: {'p': p2, 'v': v2, 'T': T2},
                3: {'p': p3, 'v': v3, 'T': T3},
                4: {'p': p4, 'v': v4, 'T': T4}
            },
            'gamma': gamma
        }
    
    def generate_pv_diagram(self, cycle_data, n_points=100):
        """Generate P-V diagram data for the diesel cycle"""
        states = cycle_data['states']
        gamma = cycle_data['gamma']
        
        # Process 1-2: Isentropic compression
        v_12 = np.linspace(states[1]['v'], states[2]['v'], n_points)
        p_12 = states[1]['p'] * (states[1]['v']/v_12)**gamma
        
        # Process 2-3: Constant pressure heat addition
        v_23 = np.linspace(states[2]['v'], states[3]['v'], n_points)
        p_23 = np.full_like(v_23, states[2]['p'])
        
        # Process 3-4: Isentropic expansion
        v_34 = np.linspace(states[3]['v'], states[4]['v'], n_points)
        p_34 = states[3]['p'] * (states[3]['v']/v_34)**gamma
        
        # Process 4-1: Constant volume heat rejection
        v_41 = np.full(n_points, states[4]['v'])
        p_41 = np.linspace(states[4]['p'], states[1]['p'], n_points)
        
        return {
            'compression': (v_12, p_12),
            'heat_addition': (v_23, p_23),
            'expansion': (v_34, p_34),
            'heat_rejection': (v_41, p_41)
        }
    
    def calculate_efficiency(self, cycle_data):
        """Calculate thermal efficiency of the diesel cycle"""
        states = cycle_data['states']
        gamma = cycle_data['gamma']
        
        # Cut-off ratio
        rho = states[3]['v'] / states[2]['v']
        
        # Diesel cycle efficiency formula
        efficiency = 1 - (1/self.compression_ratio**(gamma-1)) * (rho**gamma - 1)/(gamma * (rho - 1))
        
        return efficiency
    
    def plot_pv_diagram(self, cycle_data, title="Diesel Cycle P-V Diagram"):
        """Plot the P-V diagram"""
        pv_data = self.generate_pv_diagram(cycle_data)
        
        plt.figure(figsize=(10, 8))
        
        # Plot each process
        v_comp, p_comp = pv_data['compression']
        v_heat_add, p_heat_add = pv_data['heat_addition'] 
        v_exp, p_exp = pv_data['expansion']
        v_heat_rej, p_heat_rej = pv_data['heat_rejection']
        
        plt.plot(v_comp, p_comp, 'b-', linewidth=2, label='1→2 Compression')
        plt.plot(v_heat_add, p_heat_add, 'r-', linewidth=2, label='2→3 Heat Addition')
        plt.plot(v_exp, p_exp, 'g-', linewidth=2, label='3→4 Expansion') 
        plt.plot(v_heat_rej, p_heat_rej, 'orange', linewidth=2, label='4→1 Heat Rejection')
        
        # Mark state points
        states = cycle_data['states']
        for i in [1, 2, 3, 4]:
            plt.plot(states[i]['v'], states[i]['p'], 'ko', markersize=8)
            plt.annotate(f'{i}', (states[i]['v'], states[i]['p']), 
                        xytext=(5, 5), textcoords='offset points', fontsize=12)
        
        plt.xlabel('Volume (m³)', fontsize=12)
        plt.ylabel('Pressure (kPa)', fontsize=12)
        plt.title(title, fontsize=14)
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.ticklabel_format(style='sci', axis='both', scilimits=(0,0))
        
        # Calculate and display efficiency
        efficiency = self.calculate_efficiency(cycle_data)
        plt.text(0.05, 0.95, f'Thermal Efficiency: {efficiency:.1%}', 
                transform=plt.gca().transAxes, fontsize=12,
                bbox=dict(boxstyle="round,pad=0.3", facecolor="lightblue"))
        
        plt.tight_layout()
        plt.show()
        
        return efficiency

# Example usage
def main():
    # Engine parameters
    bore = 0.1          # 10 cm bore
    stroke = 0.1        # 10 cm stroke  
    rod_length = 0.15   # 15 cm connecting rod
    compression_ratio = 18  # Typical diesel compression ratio
    
    # Thermodynamic conditions
    p1 = 101.3          # Initial pressure (kPa)
    T1 = 300            # Initial temperature (K)
    T3 = 2200           # Peak combustion temperature (K)
    
    # Create simulator
    simulator = DieselCycleSimulator(bore, stroke, rod_length, compression_ratio)
    
    # Calculate ideal cycle
    cycle = simulator.ideal_diesel_cycle(p1, T1, T3)
    
    # Plot P-V diagram and get efficiency
    efficiency = simulator.plot_pv_diagram(cycle)
    
    # Print cycle data
    print("\nDiesel Cycle Analysis:")
    print(f"Compression Ratio: {compression_ratio}")
    print(f"Thermal Efficiency: {efficiency:.1%}")
    print("\nState Points:")
    for state, data in cycle['states'].items():
        print(f"State {state}: P={data['p']:.1f} kPa, V={data['v']*1e6:.2f} cm³, T={data['T']:.1f} K")

if __name__ == "__main__":
    main()
```

```python
import numpy as np
from numpy import pi, sin, cos, sqrt
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg
import tkinter as tk
from tkinter import ttk

# --- Thermodynamic & Kinematic Formulas ---

def volume(d, s, l, r, theta):
    """
    Computes cylinder volume at a specific crank angle (theta) 
    using standard internal combustion engine slider-crank geometry.
    """
    Vs = (pi / 4) * d**2 * s  # Swept volume
    Vc = Vs / (r - 1)         # Clearance volume
    
    # R is the ratio of connecting rod length (l) to crank radius (a = s/2)
    R_ratio = (2 * l / s)
    
    term2 = 1 + R_ratio - cos(theta)
    term3 = sqrt(R_ratio**2 - (sin(theta))**2)
    
    # Calculate instant volume based on clearance volume and crank factors
    V = Vc * (1 + 0.5 * (r - 1) * (term2 - term3))
    return V

# --- Main Dashboard Application Window ---

class DieselSimulatorApp:
    def __init__(self, root):
        self.root = root
        self.root.title("Interactive Diesel Cycle Dashboard")
        
        # --- Default Parameter States ---
        self.d = 0.1     # Bore (m)
        self.s = 0.1     # Stroke (m)
        self.l = 0.15    # Connecting rod length (m)
        self.r = 12.0    # Compression ratio
        self.p1 = 101.3  # Initial pressure (kPa)
        self.t1 = 300.0  # Initial temperature (K)
        self.gamma = 1.4 # Specific heat ratio
        self.t3 = 2500.0 # Maximum combustion temperature (K)
        
        # Container frame splitting control panels from visual charts
        self.setup_layout()
        self.calculate_cycle_data()
        self.setup_plot_and_animation()
        
    def setup_layout(self):
        """Builds control layout boxes using Tkinter TTK themed widgets."""
        control_frame = ttk.LabelFrame(self.root, text=" Engine Configuration ")
        control_frame.pack(side=tk.LEFT, fill=tk.Y, padx=10, pady=10)
        
        # Compression Ratio Slider Interactivity Hook
        ttk.Label(control_frame, text="Compression Ratio (r):").pack(anchor=tk.W, padx=5, pady=2)
        
        # Setup the slider framework
        self.r_slider = ttk.Scale(control_frame, from_=10, to=22, orient=tk.HORIZONTAL, command=self.update_parameters)
        self.r_slider.pack(fill=tk.X, padx=5, pady=5)
        
        # Dynamic readout showing compression metric properties instantly
        self.r_label = ttk.Label(control_frame, text=f"Value: {self.r:.1f}")
        self.r_label.pack(anchor=tk.W, padx=5, pady=2)
        
        # Safe trigger invocation initialization AFTER label exists to avoid race-condition AttributeError crashes
        self.r_slider.set(self.r)
        
        # Graph presentation frame area mapping inside Tk framework windows
        self.plot_frame = ttk.Frame(self.root)
        self.plot_frame.pack(side=tk.RIGHT, fill=tk.BOTH, expand=True, padx=10, pady=10)

    def calculate_cycle_data(self):
        """Executes the thermodynamic equations to produce PV arrays."""
        Vs = (pi / 4) * self.s * self.d**2
        Vc = Vs / (self.r - 1)
        
        self.v1 = Vs + Vc
        self.v2 = Vc
        
        # Process 1 -> 2: Compression calculations
        self.p2 = self.p1 * (self.r)**self.gamma   
        t2 = self.t1 * self.r**(self.gamma - 1)
        
        # Process 2 -> 3: Isobaric ignition phase calculations
        self.p3 = self.p2
        self.v3 = self.v2 * self.t3 / t2
        
        # Solve exact mechanical cutoff crank angle intersection threshold 
        self.theta_cutoff = 0
        while self.theta_cutoff < pi:
            self.theta_cutoff += 0.001
            v_check = volume(self.d, self.s, self.l, self.r, self.theta_cutoff)
            if abs(v_check - self.v3) < (0.001 * self.v3):
                break
                
        # Process 3 -> 4: Expansion trace calculations
        self.v4 = self.v1
        self.p4 = self.p3 * (self.v3 / self.v4)**self.gamma

        # Step-by-step array profile data assembly for complete background plotting traces
        self.theta_comp = np.linspace(0, pi, 100)
        self.V_comp = volume(self.d, self.s, self.l, self.r, self.theta_comp)
        self.P_comp = (self.p1 * self.v1**self.gamma) / self.V_comp**self.gamma
        
        self.theta_exp = np.linspace(self.theta_cutoff, pi, 100)
        self.V_exp = volume(self.d, self.s, self.l, self.r, self.theta_exp)
        self.P_exp = (self.p3 * self.v3**self.gamma) / self.V_exp**self.gamma
        
        # Flattening dynamic operations chronologically to cleanly feed into the animation engine loop
        # Mapping standard operational rotation sequence: Compression (0 to Pi) -> Expansion (Cutoff to Pi)
        self.anim_volumes = np.concatenate([self.V_comp, [self.v2, self.v3], self.V_exp, [self.v4, self.v1]])
        self.anim_pressures = np.concatenate([self.P_comp, [self.p2, self.p3], self.P_exp, [self.p4, self.p1]])

    def setup_plot_and_animation(self):
        """Initialises Matplotlib graphics and embeds the window canvas context into Tkinter."""
        self.fig, self.ax = plt.subplots(figsize=(6, 5))
        
        # Static baseline plotting lines structural paths configuration
        self.ax.plot(self.V_comp, self.P_comp, 'b--', alpha=0.4, label='Compression')
        self.ax.plot([self.v2, self.v3], [self.p2, self.p3], 'r--', alpha=0.4, label='Combustion')
        self.ax.plot(self.V_exp, self.P_exp, 'g--', alpha=0.4, label='Expansion')
        self.ax.plot([self.v4, self.v1], [self.p4, self.p1], 'k--', alpha=0.4, label='Exhaust')
        
        # Animated dynamic tracking point marking current air-fuel charge system state variables
        self.tracer_dot, = self.ax.plot([], [], 'ro', markersize=9, label="Engine Piston Head Tracking")
        
        # Format aesthetics styling controls setup 
        self.ax.set_title("Real-Time Diesel Cycle Animation Trace", fontsize=12)
        self.ax.set_xlabel("Volume ($m^3$)")
        self.ax.set_ylabel("Pressure ($kPa$)")
        self.ax.ticklabel_format(style='sci', axis='both', scilimits=(0,0))
        self.ax.grid(True, linestyle=":", alpha=0.6)
        self.ax.legend(loc='upper right')
        
        # Mount the Matplotlib window canvas inside the dedicated Tk layout frame frame wrapper space
        self.canvas = FigureCanvasTkAgg(self.fig, master=self.plot_frame)
        self.canvas.get_tk_widget().pack(fill=tk.BOTH, expand=True)
        
        # Run animation loop utilizing 40 milliseconds updates intervals per step
        self.ani = FuncAnimation(self.fig, self.animate_step, frames=len(self.anim_volumes), interval=40, blit=True)

    def animate_step(self, frame_idx):
        """Updates indicator coordinate placements systematically inside the running looping engine context frame instances."""
        x_val = self.anim_volumes[frame_idx]
        y_val = self.anim_pressures[frame_idx]
        self.tracer_dot.set_data([x_val], [y_val]) 
        return self.tracer_dot,

    def update_parameters(self, event=None):
        """Event callback engine processing compression ratio adjustments on slider movements events."""
        # Prevent premature callbacks execution loop triggers before UI objects are finalized
        if not hasattr(self, 'r_label'):
            return

        self.r = float(self.r_slider.get())
        self.r_label.config(text=f"Value: {self.r:.1f}")
        
        # Recalculate thermodynamics boundaries matrices profiles variations limits safely array definitions 
        self.calculate_cycle_data()
        
        # Clear out current display matrices frames boundaries data axes and redraw base traces 
        self.ax.clear()
        self.ax.plot(self.V_comp, self.P_comp, 'b--', alpha=0.4, label='Compression')
        self.ax.plot([self.v2, self.v3], [self.p2, self.p3], 'r--', alpha=0.4, label='Combustion')
        self.ax.plot(self.V_exp, self.P_exp, 'g--', alpha=0.4, label='Expansion')
        self.ax.plot([self.v4, self.v1], [self.p4, self.p1], 'k--', alpha=0.4, label='Exhaust')
        
        # Remount elements tracking properties metrics traces 
        self.tracer_dot, = self.ax.plot([], [], 'ro', markersize=9, label="Engine Piston Head Tracking")
        self.ax.set_title("Real-Time Diesel Cycle Animation Trace", fontsize=12)
        self.ax.set_xlabel("Volume ($m^3$)")
        self.ax.set_ylabel("Pressure ($kPa$)")
        self.ax.ticklabel_format(style='sci', axis='both', scilimits=(0,0))
        self.ax.legend(loc='upper right')
        self.ax.grid(True, linestyle=":", alpha=0.6)

# --- Entry Initiation Execution Main Process Target Blocks ---
if __name__ == "__main__":
    root = tk.Tk()
    app = DieselSimulatorApp(root)
    root.mainloop()
```

```python
import numpy as np
from numpy import pi, sin, cos, sqrt

import tkinter as tk
from tkinter import ttk

import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg


# ==========================================================
# Engine Geometry Function
# ==========================================================

def volume(d, s, l, r, theta):
    """
    Cylinder volume from slider-crank mechanism.

    Parameters
    ----------
    d : Bore (m)
    s : Stroke (m)
    l : Connecting rod length (m)
    r : Compression ratio
    theta : Crank angle (rad)

    Returns
    -------
    Volume (m³)
    """

    Vs = (pi / 4) * d**2 * s
    Vc = Vs / (r - 1)

    a = s / 2
    R = l / a

    V = Vc * (
        1
        + 0.5 * (r - 1)
        * (
            R + 1
            - np.cos(theta)
            - np.sqrt(R**2 - np.sin(theta) ** 2)
        )
    )

    return V


# ==========================================================
# Diesel Cycle Simulator
# ==========================================================

class DieselSimulatorApp:

    def __init__(self, root):

        self.root = root
        self.root.title("Interactive Diesel Cycle Dashboard")
        self.root.geometry("1200x700")

        # ----------------------------
        # Engine Parameters
        # ----------------------------
        self.d = 0.10          # Bore (m)
        self.s = 0.10          # Stroke (m)
        self.l = 0.15          # Connecting Rod (m)
        self.r = 12.0          # Compression Ratio

        self.p1 = 101.3        # kPa
        self.t1 = 300.0        # K
        self.gamma = 1.4

        self.t3 = 2500.0       # Peak temperature

        # Animation state
        self.is_playing = True
        self.current_frame = 0

        # Build UI
        self.setup_layout()

        # Compute initial cycle
        self.calculate_cycle_data()

        # Create plot
        self.setup_plot_and_animation()

        # Initialize sliders
        self.r_slider.set(self.r)
        self.d_slider.set(self.d)
        self.s_slider.set(self.s)
        self.t3_slider.set(self.t3)

    # ======================================================
    # UI
    # ======================================================

    def setup_layout(self):

        control_frame = ttk.LabelFrame(
            self.root,
            text=" Engine Configuration "
        )

        control_frame.pack(
            side=tk.LEFT,
            fill=tk.Y,
            padx=10,
            pady=10
        )

        # Compression Ratio
        ttk.Label(
            control_frame,
            text="Compression Ratio (r)"
        ).pack(anchor=tk.W, padx=5)

        self.r_slider = ttk.Scale(
            control_frame,
            from_=10,
            to=22,
            orient=tk.HORIZONTAL,
            command=self.update_parameters
        )

        self.r_slider.pack(fill=tk.X, padx=5)

        self.r_label = ttk.Label(
            control_frame,
            text=f"Value: {self.r:.1f}"
        )

        self.r_label.pack(anchor=tk.W, padx=5, pady=3)

        # Bore
        ttk.Label(
            control_frame,
            text="Bore (d) [m]"
        ).pack(anchor=tk.W, padx=5)

        self.d_slider = ttk.Scale(
            control_frame,
            from_=0.05,
            to=0.15,
            orient=tk.HORIZONTAL,
            command=self.update_parameters
        )

        self.d_slider.pack(fill=tk.X, padx=5)

        self.d_label = ttk.Label(
            control_frame,
            text=f"Value: {self.d:.3f} m"
        )

        self.d_label.pack(anchor=tk.W, padx=5, pady=3)

        # Stroke
        ttk.Label(
            control_frame,
            text="Stroke (s) [m]"
        ).pack(anchor=tk.W, padx=5)

        self.s_slider = ttk.Scale(
            control_frame,
            from_=0.05,
            to=0.15,
            orient=tk.HORIZONTAL,
            command=self.update_parameters
        )

        self.s_slider.pack(fill=tk.X, padx=5)

        self.s_label = ttk.Label(
            control_frame,
            text=f"Value: {self.s:.3f} m"
        )

        self.s_label.pack(anchor=tk.W, padx=5, pady=3)

        # Maximum Temperature
        ttk.Label(
            control_frame,
            text="Max Combustion Temp (T3) [K]"
        ).pack(anchor=tk.W, padx=5)

        self.t3_slider = ttk.Scale(
            control_frame,
            from_=1800,
            to=3000,
            orient=tk.HORIZONTAL,
            command=self.update_parameters
        )

        self.t3_slider.pack(fill=tk.X, padx=5)

        self.t3_label = ttk.Label(
            control_frame,
            text=f"Value: {self.t3:.0f} K"
        )

        self.t3_label.pack(anchor=tk.W, padx=5, pady=3)

        ttk.Separator(control_frame).pack(
            fill=tk.X,
            padx=5,
            pady=10
        )

        self.play_btn = ttk.Button(
            control_frame,
            text="Pause Animation",
            command=self.toggle_animation
        )

        self.play_btn.pack(
            fill=tk.X,
            padx=5,
            pady=5
        )

        # Plot frame
        self.plot_frame = ttk.Frame(self.root)

        self.plot_frame.pack(
            side=tk.RIGHT,
            fill=tk.BOTH,
            expand=True,
            padx=10,
            pady=10
        )

    # ======================================================
    # Thermodynamics
    # ======================================================

    def calculate_cycle_data(self):

        Vs = (pi / 4) * self.d**2 * self.s
        Vc = Vs / (self.r - 1)

        self.v1 = Vs + Vc
        self.v2 = Vc

        # Compression
        self.p2 = self.p1 * (self.r ** self.gamma)
        t2 = self.t1 * (self.r ** (self.gamma - 1))

        # Constant Pressure Heat Addition
        self.p3 = self.p2

        self.v3 = self.v2 * self.t3 / t2

        if self.v3 > self.v1:
            self.v3 = self.v1

        # Find cutoff angle
        self.theta_cutoff = 0.0

        while self.theta_cutoff < pi:

            self.theta_cutoff += 0.001

            v_check = volume(
                self.d,
                self.s,
                self.l,
                self.r,
                self.theta_cutoff
            )

            if abs(v_check - self.v3) < (0.001 * self.v3):
                break

        # Expansion
        self.v4 = self.v1

        self.p4 = self.p3 * (
            self.v3 / self.v4
        ) ** self.gamma

        # Compression Curve

        self.theta_comp = np.linspace(
            0,
            pi,
            200
        )

        self.V_comp = volume(
            self.d,
            self.s,
            self.l,
            self.r,
            self.theta_comp
        )

        self.P_comp = (
            self.p1 * self.v1**self.gamma
        ) / self.V_comp**self.gamma

        # Expansion Curve

        self.theta_exp = np.linspace(
            self.theta_cutoff,
            pi,
            200
        )

        self.V_exp = volume(
            self.d,
            self.s,
            self.l,
            self.r,
            self.theta_exp
        )

        self.P_exp = (
            self.p3 * self.v3**self.gamma
        ) / self.V_exp**self.gamma

        self.anim_volumes = np.concatenate(
            [
                self.V_comp,
                [self.v2, self.v3],
                self.V_exp,
                [self.v4, self.v1]
            ]
        )

        self.anim_pressures = np.concatenate(
            [
                self.P_comp,
                [self.p2, self.p3],
                self.P_exp,
                [self.p4, self.p1]
            ]
        )

    # ======================================================
    # Plot Setup
    # ======================================================

    def setup_plot_and_animation(self):

        self.fig, self.ax = plt.subplots(
            figsize=(7, 6)
        )

        self.lines_comp, = self.ax.plot(
            self.V_comp,
            self.P_comp,
            'b--',
            label="Compression"
        )

        self.lines_comb, = self.ax.plot(
            [self.v2, self.v3],
            [self.p2, self.p3],
            'r--',
            label="Combustion"
        )

        self.lines_exp, = self.ax.plot(
            self.V_exp,
            self.P_exp,
            'g--',
            label="Expansion"
        )

        self.lines_exh, = self.ax.plot(
            [self.v4, self.v1],
            [self.p4, self.p1],
            'k--',
            label="Heat Rejection"
        )

        self.tracer_dot, = self.ax.plot(
            [],
            [],
            'ro',
            markersize=8
        )

        self.ax.set_title(
            "Real-Time Diesel Cycle P-V Diagram"
        )

        self.ax.set_xlabel(
            "Volume (m³)"
        )

        self.ax.set_ylabel(
            "Pressure (kPa)"
        )

        self.ax.grid(True)
        self.ax.legend()

        self.ax.ticklabel_format(
            style='sci',
            axis='x',
            scilimits=(0, 0)
        )

        self.autoscale_axes()

        self.fig.tight_layout()

        self.canvas = FigureCanvasTkAgg(
            self.fig,
            master=self.plot_frame
        )

        self.canvas.draw()

        self.canvas.get_tk_widget().pack(
            fill=tk.BOTH,
            expand=True
        )

        self.ani = FuncAnimation(
            self.fig,
            self.animate_step,
            frames=self.gen_frames(),
            interval=40,
            blit=False,
            cache_frame_data=False
        )

    # ======================================================
    # Animation
    # ======================================================

    def gen_frames(self):

        while True:

            yield self.current_frame

            if self.is_playing:
                self.current_frame = (
                    self.current_frame + 1
                ) % len(self.anim_volumes)

    def animate_step(self, frame_index):

        x = self.anim_volumes[frame_index]
        y = self.anim_pressures[frame_index]

        self.tracer_dot.set_data(
            [x],
            [y]
        )

        return self.tracer_dot,

    # ======================================================
    # Controls
    # ======================================================

    def toggle_animation(self):

        self.is_playing = not self.is_playing

        if self.is_playing:
            self.play_btn.config(
                text="Pause Animation"
            )
        else:
            self.play_btn.config(
                text="Play Animation"
            )

    def autoscale_axes(self):

        max_vol = np.max(self.anim_volumes)
        min_vol = np.min(self.anim_volumes)

        max_p = np.max(self.anim_pressures)
        min_p = np.min(self.anim_pressures)

        vol_margin = (max_vol - min_vol) * 0.1
        p_margin = (max_p - min_p) * 0.1

        self.ax.set_xlim(
            min_vol - vol_margin,
            max_vol + vol_margin
        )

        self.ax.set_ylim(
            min_p - p_margin,
            max_p + p_margin
        )

    # ======================================================
    # Slider Updates
    # ======================================================

    def update_parameters(self, event=None):

        if not hasattr(self, 'canvas'):
            return

        self.r = float(self.r_slider.get())
        self.d = float(self.d_slider.get())
        self.s = float(self.s_slider.get())
        self.t3 = float(self.t3_slider.get())

        self.r_label.config(
            text=f"Value: {self.r:.1f}"
        )

        self.d_label.config(
            text=f"Value: {self.d:.3f} m"
        )

        self.s_label.config(
            text=f"Value: {self.s:.3f} m"
        )

        self.t3_label.config(
            text=f"Value: {self.t3:.0f} K"
        )

        self.calculate_cycle_data()

        self.lines_comp.set_data(
            self.V_comp,
            self.P_comp
        )

        self.lines_comb.set_data(
            [self.v2, self.v3],
            [self.p2, self.p3]
        )

        self.lines_exp.set_data(
            self.V_exp,
            self.P_exp
        )

        self.lines_exh.set_data(
            [self.v4, self.v1],
            [self.p4, self.p1]
        )

        self.autoscale_axes()

        if self.current_frame >= len(self.anim_volumes):
            self.current_frame = 0

        self.canvas.draw_idle()


# ==========================================================
# Main Program
# ==========================================================

if __name__ == "__main__":

    root = tk.Tk()

    app = DieselSimulatorApp(root)

    root.mainloop()
```
