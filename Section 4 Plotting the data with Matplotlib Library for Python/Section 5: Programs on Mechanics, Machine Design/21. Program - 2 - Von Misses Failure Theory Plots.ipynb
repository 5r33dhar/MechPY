```python
#AI NW
import tkinter as tk
from tkinter import ttk
import numpy as np
import matplotlib.pyplot as plt
from numpy import sin, cos, pi
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg

# ==========================================
# PRESET CALCULATIONS & ENGINE HANDLERS
# ==========================================
def on_preset_change(event):
    """Handles logic when the user picks a preset option from the Combobox dropdown menu."""
    selection = combo_preset.get()
    Syt = float(slider_syt.get())
    
    # Temporarily remove trace hooks to prevent slider recursion loops during override updates
    slider_s1.config(command="")
    slider_s2.config(command="")
    
    if selection == "Uniaxial Tension":
        slider_s1.set(int(Syt * 0.5))
        slider_s2.set(0)
    elif selection == "Pure Shear":
        slider_s1.set(int(Syt * 0.5))
        slider_s2.set(int(-Syt * 0.5))
    elif selection == "Biaxial Equal Tension":
        slider_s1.set(int(Syt * 0.6))
        slider_s2.set(int(Syt * 0.6))
    
    # Restore standard canvas updates triggers to scale sliders
    slider_s1.config(command=on_slider_move)
    slider_s2.config(command=on_slider_move)
    
    # Repaint framework coordinates layout metrics frame
    update_dashboard()

def on_canvas_click(event):
    """Allows users to click right on the plot canvas plane to set coordinates instantly."""
    # Guard clause to verify clicks occurred inside valid plotting margins
    if event.xdata is None or event.ydata is None:
        return
        
    # Jump selection preset indicator value index back to custom tracking
    combo_preset.set("Custom (Interactive / Click Canvas)")
    
    # Temporarily separate command paths during value overrides
    slider_s1.config(command="")
    slider_s2.config(command="")
    
    # Force sliders track matching data points
    slider_s1.set(int(event.xdata))
    slider_s2.set(int(event.ydata))
    
    # Restore tracking callback loops
    slider_s1.config(command=on_slider_move)
    slider_s2.config(command=on_slider_move)
    
    update_dashboard()

def on_slider_move(*args):
    """Intercepts slider track changes, setting dropdown state to custom mode."""
    if combo_preset.get() != "Custom (Interactive / Click Canvas)":
        combo_preset.set("Custom (Interactive / Click Canvas)")
    update_dashboard()

# ==========================================
# DYNAMIC REAL-TIME GRAPHICS & MATH CORE
# ==========================================
def update_dashboard(*args):
    """
    Callback function triggered automatically on any slider movement.
    Handles engineering math, dynamic safety calculations, and redraws the plot.
    """
    # 1. FETCH ENTRY PARAMETERS SAFELY FROM TKINTER SLIDERS
    try:
        Syt = float(slider_syt.get())  # Yield strength of the chosen material (MPa)
        s1 = float(slider_s1.get())    # First principal stress sigma_1 (MPa)
        s2 = float(slider_s2.get())    # Second principal stress sigma_2 (MPa)
    except Exception:
        return  # Safeguard against rendering anomalies if values are briefly empty

    # Update numerical label readout elements next to each slider track
    lbl_syt_val.config(text=f"{int(Syt)} MPa")
    lbl_s1_val.config(text=f"{int(s1)} MPa")
    lbl_s2_val.config(text=f"{int(s2)} MPa")

    # 2. MECHANICS OF MATERIALS CALCULATIONS (YIELD THEORIES)
    # Von Mises Equivalent Stress: Distortion Energy Theory
    von_mises_eq = np.sqrt(s1**2 - (s1 * s2) + s2**2)
    
    # Tresca Equivalent Stress: Maximum Shear Stress (MSS) Theory
    tresca_eq = max(abs(s1), abs(s2), abs(s1 - s2))

    # Calculate Factor of Safety (FoS = Syt / Equivalent Stress)
    fos_mises = Syt / von_mises_eq if von_mises_eq > 0 else float('inf')
    fos_tresca = Syt / tresca_eq if tresca_eq > 0 else float('inf')

    # Update dashboard headline status indicators with the computed equivalent values
    lbl_vme_val.config(text=f"{von_mises_eq:.1f} MPa")
    lbl_tre_val.config(text=f"{tresca_eq:.1f} MPa")
    
    # Color FoS readout text dynamically: Green if safe (both FoS >= 1.0), Red if structural failure occurs
    is_safe = (fos_mises >= 1.0 and fos_tresca >= 1.0)
    fos_color = "#4CAF50" if is_safe else "#E91E63"
    
    # Format FoS numbers to clip at 99+ value ranges instead of extending out infinitely
    mises_str = f"{fos_mises:.2f}" if fos_mises < 99 else "99.00+"
    tresca_str = f"{fos_tresca:.2f}" if fos_tresca < 99 else "99.00+"
    lbl_fos_val.config(text=f"{mises_str} / {tresca_str}", foreground=fos_color)

    # 3. GRAPH PLOTTING RENDER PIPELINE
    ax.clear()  # Wipe out the previous frame paths to prevent overlay memory leaks

    # Calculate geometric semi-axes dimensions for the standard Von Mises Ellipse
    a = np.sqrt(2) * Syt          # Long boundary diameter projection
    b = np.sqrt(2 / 3) * Syt      # Short boundary diameter projection

    # Trace out the foundational unrotated ellipse coordinates using polar math
    theta = np.linspace(0, 2 * pi, 720)
    x = a * cos(theta)
    y = b * sin(theta)

    # Apply a 45-degree (pi/4 radian) rotation matrix transformation to match the plane coordinate system
    rot_ang = pi / 4
    xd = x * cos(rot_ang) - y * sin(rot_ang)
    yd = x * sin(rot_ang) + y * cos(rot_ang)

    # Define absolute polygon vertices for the Tresca structural failure envelope (Hexagon)
    x_ShearPoints = [Syt, Syt, 0, -Syt, -Syt, 0, Syt]
    y_ShearPoints = [0, Syt, Syt, 0, -Syt, -Syt, 0]

    # Render primary failure boundary envelopes onto the plane coordinate space
    ax.plot(xd, yd, color='#1E88E5', linewidth=2, label='Von Mises Envelope (Yield)')
    ax.plot(x_ShearPoints, y_ShearPoints, color='#E53935', linewidth=1.5, linestyle='--', label='Tresca Envelope (Max Shear)')

    # Build baseline horizontal and vertical quadrant center dividing lines
    ax.axhline(0, color='#374151', linewidth=1)
    ax.axvline(0, color='#374151', linewidth=1)

    # Select dynamic point marker coloration (Matches target safety envelope criteria state)
    point_color = '#4CAF50' if is_safe else '#E53935'
    ax.scatter([s1], [s2], color=point_color, s=80, zorder=5, label='Current Stress State ($\sigma_1$, $\sigma_2$)')
    
    # Render a low-contrast "Safe Zone" label watermark inside the failure boundaries if material is safe
    if is_safe:
        ax.text(0, -50, "Safe Zone", color='#4B5563', fontsize=10, ha='center', va='center')

    # Formatting limits and modern low-contrast aesthetic tracking rule overrides
    ax.set_xlim(-500, 500)
    ax.set_ylim(-500, 500)
    ax.set_xlabel(r'$\sigma_1$ (MPa)', color='#9CA3AF', labelpad=10)
    ax.set_ylabel(r'$\sigma_2$ (MPa)', color='#9CA3AF', labelpad=10)
    
    # Orient position labels neatly across the top and right margins like the reference layout image
    ax.xaxis.set_label_position('top')
    ax.yaxis.set_label_position('right')
    
    # Format grid details, ticks, and ensure absolute 1:1 scale aspect parity for accurate geometry
    ax.tick_params(colors='#9CA3AF', labelsize=9)
    ax.grid(True, color='#1F2937', linestyle=':', alpha=0.6)
    ax.set_aspect('equal', adjustable='box')

    # Signal the connected Tkinter parent element to immediately repaint the backend layout frame
    canvas.draw()


# ==========================================
# DESKTOP USER INTERFACE NATIVE WINDOW SETUP
# ==========================================
root = tk.Tk()
root.title("Failure Criteria Dashboard")
root.geometry("680x880")
root.configure(bg="#1E222B")  # Matte dark charcoal visual palette skin background

# Establish modern widget styling patterns (Clam engine structure configurations)
style = ttk.Style()
style.theme_use('clam')
style.configure('.', background="#1E222B", foreground="#FFFFFF")
style.configure('TLabel', background="#1E222B", foreground="#FFFFFF", font=('Arial', 10))
style.configure('Heading.TLabel', background="#1E222B", foreground="#9CA3AF", font=('Arial', 10, 'bold'))
style.configure('Value.TLabel', background="#1E222B", foreground="#FFFFFF", font=('Arial', 12, 'bold'))

# Clean dark drop-down combo styling states implementation (Fixed syntax cut-off)
style.configure('TCombobox', fieldbackground="#2D3139", background="#1E222B", foreground="#FFFFFF", arrowcolor="#FFFFFF")
style.map('TCombobox', fieldbackground=[('readonly', '#2D3139')], foreground=[('readonly', '#FFFFFF')])

# 1. MAIN INDICATOR BADGES GRID (TOP DASHBOARD HEADER PANEL)
frame_legend = tk.Frame(root, bg="#1E222B")
frame_legend.pack(fill=tk.X, padx=20, pady=10)

# Von Mises dynamic numeric display output module
lbl_vme_title = ttk.Label(frame_legend, text="Von Mises Eq.", style='Heading.TLabel')
lbl_vme_title.grid(row=0, column=0, padx=20, pady=2)
lbl_vme_val = ttk.Label(frame_legend, text="0.0 MPa", style='Value.TLabel')
lbl_vme_val.grid(row=1, column=0, padx=20, pady=2)

# Tresca dynamic numeric display output module
lbl_tre_title = ttk.Label(frame_legend, text="Tresca Eq.", style='Heading.TLabel')
lbl_tre_title.grid(row=0, column=1, padx=20, pady=2)
lbl_tre_val = ttk.Label(frame_legend, text="0.0 MPa", style='Value.TLabel')
lbl_tre_val.grid(row=1, column=1, padx=20, pady=2)

# Combined Factor of Safety readout display output module
lbl_fos_title = ttk.Label(frame_legend, text="FoS (Mises/Tresca)", style='Heading.TLabel')
lbl_fos_title.grid(row=0, column=2, padx=20, pady=2)
lbl_fos_val = ttk.Label(frame_legend, text="0.00 / 0.00", style='Value.TLabel')
lbl_fos_val.grid(row=1, column=2, padx=20, pady=2)

# Standardize alignment weights across header cells evenly
frame_legend.grid_columnconfigure((0, 1, 2), weight=1)

# 2. EMBEDDED MATPLOTLIB GRAPH SURFACE ATTACHMENT
fig, ax = plt.subplots(figsize=(6, 5.2), facecolor='#1E222B')
ax.set_facecolor('#1E222B')

# Strip away traditional white exterior bounding charts outlines for modern minimalism
for spine in ax.spines.values():
    spine.set_visible(False)

# Link the customized plot figure securely into the parent Tkinter framework widget flow tree
```
