```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.misc import derivative

def f(x):
    return x ** 3 - x - 1

def g(x):
    # Using a smaller dx or an explicit derivative function is preferred
    return derivative(f, x, dx=1e-4)

# 1. Choose a safer initial guess outside the local extrema region
x0 = 1.5 
error = 1e-8
max_iter = 100
iteration = 0

# 2. Use abs() to ensure we check closeness to zero from both sides
while abs(f(x0)) > error and iteration < max_iter:
    x1 = x0 - (f(x0) / g(x0))
    x0 = x1
    iteration += 1

print(f"Root found: {x0} in {iteration} iterations")

# Plotting the function to visualize the root
x = np.linspace(-2, 2, 100)
plt.plot(x, f(x), label='$f(x) = x^3 - x - 1$')
plt.axhline(0, color='black', lw=0.5, ls='--')
plt.axvline(x0, color='red', lw=0.5, ls='--', label=f'Root $\\approx$ {x0:.4f}')
plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import approx_fprime  # Modern replacement

def f(x):
    # approx_fprime expects a float array, so we ensure x behaves nicely
    return float(x) ** 3 - float(x) - 1

def g(x):
    # Returns the numerical gradient at a specific point
    # [1e-4] represents the step size (epsilon) for calculation
    return approx_fprime(np.array([x]), f, [1e-4])[0]

x0 = 1.5 
error = 1e-8
max_iter = 100
iteration = 0

while abs(f(x0)) > error and iteration < max_iter:
    x1 = x0 - (f(x0) / g(x0))
    x0 = x1
    iteration += 1

print(f"Root found: {x0}")
```
