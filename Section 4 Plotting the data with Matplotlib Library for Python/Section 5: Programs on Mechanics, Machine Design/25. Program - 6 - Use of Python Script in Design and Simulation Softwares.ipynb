```python
doc = FreeCAD. newDocument ("my_design")
import Part
plate = Part.makeBox (50, 30, 2)
verticals = [edge for edge in plate.Edges if edge.BoundBox. ZLength]
plate = plate.makeFillet (5, verticals)
cyl = Part.makeCylinder (10, 2, Vector (20, 15, 0))
plate = plate.cut (cyl)
doc.addObject ("Part: :Feature", "plate") .Shape = plate
```

```python
import Adams

# 1. Create the model
m = Adams.Models.create(name='Constraints')

# 2. Create the Rigid Bodies
p1 = m.Parts.createRigidBody(name='Part_1')
p2 = m.Parts.createRigidBody(name='Part_2')

# 3. Create Markers for Part_1 and Part_2
mar1 = p1.Markers.create(name='Marker_1')
mar2 = p2.Markers.create(name='Marker_2', location=[500, 500, 0])
mar3 = p2.Markers.create(name='Marker_3', location=[500, -500, 0])
mar4 = p2.Markers.create(name='Marker_4', location=[-500, 500, 0])
mar5 = p2.Markers.create(name='Marker_5', location=[-500, -500, 0])

# 4. Create Constraints (Joints) connecting Part_1 (via mar1) to Part_2 markers
j1 = m.Constraints.createTranslational(name='Joint_1', i_marker_name='.Constraints.Part_1.Marker_1', j_marker_name='.Constraints.Part_2.Marker_2')
j2 = m.Constraints.createRevolute(name='Joint_2', i_marker_name='.Constraints.Part_1.Marker_1', j_marker_name='.Constraints.Part_2.Marker_2')
j3 = m.Constraints.createCylindrical(name='Joint_3', i_marker_name='.Constraints.Part_1.Marker_1', j_marker_name='.Constraints.Part_2.Marker_2')
j4 = m.Constraints.createUniversal(name='Joint_4', i_marker_name='.Constraints.Part_1.Marker_1', j_marker_name='.Constraints.Part_2.Marker_2')
j5 = m.Constraints.createSpherical(name='Joint_5', i_marker_name='.Constraints.Part_1.Marker_1', j_marker_name='.Constraints.Part_2.Marker_2')
j6 = m.Constraints.createPlanar(name='Joint_6', i_marker_name='.Constraints.Part_1.Marker_1', j_marker_name='.Constraints.Part_2.Marker_2')

# Note: Complex joints (Rack & Pinion, Screw, Constant Velocity) usually require specialized arguments 
# like pitch, gear ratio, or a cv_marker. Below are basic structural connections:
j7 = m.Constraints.createRackpin(name='Joint_7', i_marker_name='.Constraints.Part_1.Marker_1', j_marker_name='.Constraints.Part_2.Marker_2')
j8 = m.Constraints.createScrew(name='Joint_8', i_marker_name='.Constraints.Part_1.Marker_1', j_marker_name='.Constraints.Part_2.Marker_2', pitch=1.0)
j9 = m.Constraints.createConvel(name='Joint_9', i_marker_name='.Constraints.Part_1.Marker_1', j_marker_name='.Constraints.Part_2.Marker_2')
```
