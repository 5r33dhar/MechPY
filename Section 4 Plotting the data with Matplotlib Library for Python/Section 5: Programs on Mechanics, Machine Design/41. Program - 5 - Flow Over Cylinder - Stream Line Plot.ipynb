```python
#AI
import numpy as np
import matplotlib.pyplot as plt
from numpy import sin, cos

# 1. Grid Setup
x = np.linspace(-5, 5, 300)
y = np.linspace(-5, 5, 300)
X, Y = np.meshgrid(x, y)

# 2. Dynamic Flow Parameters (Modify these to change the cylinder)
U = 1.0          # Uniform upstream flow velocity
R = 1.2          # New custom radius of the cylinder
x0, y0 = -1.0, 0.5  # New center coordinates of the shifted cylinder

# 3. Relative coordinate mapping from the new center
X_rel = X - x0
Y_rel = Y - y0

# 4. Polar coordinate conversion relative to the new center
r = np.sqrt(X_rel**2 + Y_rel**2)
theta = np.arctan2(Y_rel, X_rel)

# 5. Potential flow velocity equations
Vr = U * (1 - (R**2 / r**2)) * cos(theta)
Vt = -U * (1 + (R**2 / r**2)) * sin(theta)

# 6. Transform back to Cartesian directions
Vx = Vr * cos(theta) - Vt * sin(theta)
Vy = Vr * sin(theta) + Vt * cos(theta)

# 7. Mask out the interior of the shifted cylinder
Vx[r < R] = np.nan
Vy[r < R] = np.nan

# 8. Plotting
plt.figure(figsize=(6, 6), dpi=100)

# Draw the custom shifted cylinder
cylinder = plt.Circle((x0, y0), R, color='darkgrey', zorder=3)
plt.gca().add_patch(cylinder)

# Plot streamlines colored by velocity magnitude
v_mag = np.sqrt(Vx**2 + Vy**2)
stream = plt.streamplot(X, Y, Vx, Vy, color=v_mag, density=[1.5, 1.5], cmap='viridis', zorder=2)
plt.colorbar(stream.lines, label='Velocity Magnitude')

# Layout tuning
plt.xlim(-5, 5)
plt.ylim(-5, 5)
plt.xlabel('X')
plt.ylabel('Y')
plt.title(f'Flow Around a Cylinder (R={R}, Center=({x0},{y0}))')
plt.gca().set_aspect('equal', adjustable='box')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
import tkinter as tk
import numpy as np
import matplotlib.pyplot as plt

from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg
from matplotlib.animation import FuncAnimation


class RealTimeFlowApp:
    def __init__(self, root):
        self.root = root
        self.root.title("Animated Cylinder Flow")

        # Parameters
        self.U = 1.0
        self.R = 1.2

        # Grid
        self.x = np.linspace(-5, 5, 100)
        self.y = np.linspace(-5, 5, 100)
        self.X, self.Y = np.meshgrid(self.x, self.y)

        # Figure
        self.fig, self.ax = plt.subplots(figsize=(7, 7))

        self.canvas = FigureCanvasTkAgg(self.fig, master=root)
        self.canvas.get_tk_widget().pack(fill="both", expand=True)

        # Start animation
        self.ani = FuncAnimation(
            self.fig,
            self.update_plot,
            interval=50,
            cache_frame_data=False
        )

    def compute_flow(self, x0, y0):
        X_rel = self.X - x0
        Y_rel = self.Y - y0

        r = np.sqrt(X_rel**2 + Y_rel**2)
        r = np.where(r == 0, 1e-10, r)

        theta = np.arctan2(Y_rel, X_rel)

        Vr = self.U * (1 - (self.R**2 / r**2)) * np.cos(theta)
        Vt = -self.U * (1 + (self.R**2 / r**2)) * np.sin(theta)

        Vx = Vr * np.cos(theta) - Vt * np.sin(theta)
        Vy = Vr * np.sin(theta) + Vt * np.cos(theta)

        mask = r < self.R
        Vx[mask] = np.nan
        Vy[mask] = np.nan

        return Vx, Vy

    def update_plot(self, frame):
        self.ax.clear()

        # Moving cylinder
        x0 = -2 + 3 * np.sin(frame * 0.05)
        y0 = 1.5 * np.cos(frame * 0.05)

        Vx, Vy = self.compute_flow(x0, y0)

        velocity = np.sqrt(Vx**2 + Vy**2)

        stream = self.ax.streamplot(
            self.X,
            self.Y,
            Vx,
            Vy,
            color=velocity,
            density=1.5,
            cmap="viridis"
        )

        # Cylinder
        cylinder = plt.Circle(
            (x0, y0),
            self.R,
            color="dimgray",
            zorder=3
        )
        self.ax.add_patch(cylinder)

        self.ax.set_xlim(-5, 5)
        self.ax.set_ylim(-5, 5)
        self.ax.set_aspect("equal")

        self.ax.set_title(
            f"Flow Around Moving Cylinder\n"
            f"Center=({x0:.2f}, {y0:.2f})"
        )

        self.ax.grid(True, linestyle="--", alpha=0.4)

        self.canvas.draw_idle()


if __name__ == "__main__":
    root = tk.Tk()
    root.geometry("900x800")

    app = RealTimeFlowApp(root)

    root.mainloop()
```

```python
import tkinter as tk
import numpy as np
import matplotlib.pyplot as plt

from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg
from matplotlib.animation import FuncAnimation


class RealTimeFlowApp:

    def __init__(self, root):
        self.root = root
        self.root.title("Interactive Flow Around Cylinder")
        self.root.geometry("1000x800")

        # Flow parameters
        self.U = 1.0

        # Cylinder properties
        self.x0 = -1.0
        self.y0 = 0.5
        self.R = 1.2

        # Drag state
        self.dragging = False

        # Grid
        self.x = np.linspace(-5, 5, 120)
        self.y = np.linspace(-5, 5, 120)
        self.X, self.Y = np.meshgrid(self.x, self.y)

        # Controls Frame
        control_frame = tk.Frame(root)
        control_frame.pack(fill="x")

        # Radius Slider
        self.radius_slider = tk.Scale(
            control_frame,
            from_=0.3,
            to=2.5,
            resolution=0.1,
            orient=tk.HORIZONTAL,
            label="Cylinder Radius"
        )
        self.radius_slider.set(self.R)
        self.radius_slider.pack(fill="x", padx=10)

        # Position Label
        self.pos_label = tk.Label(
            control_frame,
            text=f"Center: ({self.x0:.2f}, {self.y0:.2f})"
        )
        self.pos_label.pack(pady=5)

        # Matplotlib Figure
        self.fig, self.ax = plt.subplots(figsize=(8, 8))

        self.canvas = FigureCanvasTkAgg(
            self.fig,
            master=root
        )
        self.canvas.get_tk_widget().pack(
            fill=tk.BOTH,
            expand=True
        )

        # Mouse Events
        self.canvas.mpl_connect(
            "button_press_event",
            self.on_press
        )

        self.canvas.mpl_connect(
            "button_release_event",
            self.on_release
        )

        self.canvas.mpl_connect(
            "motion_notify_event",
            self.on_motion
        )

        # Animation
        self.ani = FuncAnimation(
            self.fig,
            self.update_plot,
            interval=40,
            cache_frame_data=False
        )

    def compute_flow(self):

        self.R = self.radius_slider.get()

        X_rel = self.X - self.x0
        Y_rel = self.Y - self.y0

        r = np.sqrt(X_rel**2 + Y_rel**2)
        r = np.where(r == 0, 1e-10, r)

        theta = np.arctan2(Y_rel, X_rel)

        Vr = self.U * (
            1 - (self.R**2 / r**2)
        ) * np.cos(theta)

        Vt = -self.U * (
            1 + (self.R**2 / r**2)
        ) * np.sin(theta)

        Vx = (
            Vr * np.cos(theta)
            - Vt * np.sin(theta)
        )

        Vy = (
            Vr * np.sin(theta)
            + Vt * np.cos(theta)
        )

        mask = r < self.R

        Vx[mask] = np.nan
        Vy[mask] = np.nan

        return Vx, Vy

    def on_press(self, event):

        if event.xdata is None or event.ydata is None:
            return

        distance = np.sqrt(
            (event.xdata - self.x0) ** 2 +
            (event.ydata - self.y0) ** 2
        )

        if distance <= self.R:
            self.dragging = True

    def on_release(self, event):
        self.dragging = False

    def on_motion(self, event):

        if not self.dragging:
            return

        if event.xdata is None or event.ydata is None:
            return

        self.x0 = event.xdata
        self.y0 = event.ydata

        self.pos_label.config(
            text=f"Center: ({self.x0:.2f}, {self.y0:.2f})"
        )

    def update_plot(self, frame):

        self.ax.clear()

        Vx, Vy = self.compute_flow()

        velocity = np.sqrt(
            Vx**2 + Vy**2
        )

        self.ax.streamplot(
            self.X,
            self.Y,
            Vx,
            Vy,
            color=velocity,
            density=1.8,
            cmap="viridis"
        )

        cylinder = plt.Circle(
            (self.x0, self.y0),
            self.R,
            color="dimgray",
            ec="black",
            lw=2,
            zorder=3
        )

        self.ax.add_patch(cylinder)

        self.ax.set_xlim(-5, 5)
        self.ax.set_ylim(-5, 5)

        self.ax.set_aspect(
            "equal",
            adjustable="box"
        )

        self.ax.grid(
            True,
            linestyle="--",
            alpha=0.4
        )

        self.ax.set_xlabel("X")
        self.ax.set_ylabel("Y")

        self.ax.set_title(
            f"Interactive Flow Around Cylinder\n"
            f"Radius={self.R:.2f}"
        )

        return []


if __name__ == "__main__":

    root = tk.Tk()

    app = RealTimeFlowApp(root)

    root.mainloop()
```

```python
import tkinter as tk
import numpy as np
import matplotlib.pyplot as plt

from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg
from matplotlib.animation import FuncAnimation


class RealTimeFlowApp:

    def __init__(self, root):

        self.root = root
        self.root.title("Interactive Cylinder Flow")
        self.root.geometry("1100x900")

        # -----------------------------
        # Flow Parameters
        # -----------------------------
        self.U = 1.0

        # Cylinder Properties
        self.x0 = -1.0
        self.y0 = 0.5
        self.R = 1.2

        self.dragging = False

        # -----------------------------
        # Create Grid
        # -----------------------------
        self.x = np.linspace(-5, 5, 120)
        self.y = np.linspace(-5, 5, 120)
        self.X, self.Y = np.meshgrid(self.x, self.y)

        # -----------------------------
        # Control Panel
        # -----------------------------
        controls = tk.Frame(root)
        controls.pack(fill="x", padx=10, pady=5)

        # Radius Slider
        self.radius_slider = tk.Scale(
            controls,
            label="Cylinder Radius",
            from_=0.3,
            to=2.5,
            resolution=0.1,
            orient=tk.HORIZONTAL,
            command=self.slider_changed
        )
        self.radius_slider.set(self.R)
        self.radius_slider.pack(fill="x")

        # X Slider
        self.x_slider = tk.Scale(
            controls,
            label="Cylinder X Position",
            from_=-4.0,
            to=4.0,
            resolution=0.1,
            orient=tk.HORIZONTAL,
            command=self.slider_changed
        )
        self.x_slider.set(self.x0)
        self.x_slider.pack(fill="x")

        # Y Slider
        self.y_slider = tk.Scale(
            controls,
            label="Cylinder Y Position",
            from_=-4.0,
            to=4.0,
            resolution=0.1,
            orient=tk.HORIZONTAL,
            command=self.slider_changed
        )
        self.y_slider.set(self.y0)
        self.y_slider.pack(fill="x")

        # Position Label
        self.position_label = tk.Label(
            controls,
            text=f"Center: ({self.x0:.2f}, {self.y0:.2f})",
            font=("Arial", 11, "bold")
        )
        self.position_label.pack(pady=5)

        # -----------------------------
        # Matplotlib Figure
        # -----------------------------
        self.fig, self.ax = plt.subplots(figsize=(8, 8))

        self.canvas = FigureCanvasTkAgg(
            self.fig,
            master=root
        )

        self.canvas.get_tk_widget().pack(
            fill=tk.BOTH,
            expand=True
        )

        # -----------------------------
        # Mouse Events
        # -----------------------------
        self.canvas.mpl_connect(
            "button_press_event",
            self.on_press
        )

        self.canvas.mpl_connect(
            "button_release_event",
            self.on_release
        )

        self.canvas.mpl_connect(
            "motion_notify_event",
            self.on_motion
        )

        # -----------------------------
        # Animation Loop
        # -----------------------------
        self.ani = FuncAnimation(
            self.fig,
            self.update_plot,
            interval=50,
            cache_frame_data=False
        )

    # --------------------------------------------------
    # Slider Callback
    # --------------------------------------------------
    def slider_changed(self, event=None):

        self.x0 = self.x_slider.get()
        self.y0 = self.y_slider.get()
        self.R = self.radius_slider.get()

        self.position_label.config(
            text=f"Center: ({self.x0:.2f}, {self.y0:.2f})"
        )

    # --------------------------------------------------
    # Flow Solver
    # --------------------------------------------------
    def compute_flow(self):

        X_rel = self.X - self.x0
        Y_rel = self.Y - self.y0

        r = np.sqrt(X_rel**2 + Y_rel**2)

        # Avoid division by zero
        r = np.where(r == 0, 1e-10, r)

        theta = np.arctan2(Y_rel, X_rel)

        Vr = self.U * (
            1 - (self.R**2 / r**2)
        ) * np.cos(theta)

        Vt = -self.U * (
            1 + (self.R**2 / r**2)
        ) * np.sin(theta)

        Vx = (
            Vr * np.cos(theta)
            - Vt * np.sin(theta)
        )

        Vy = (
            Vr * np.sin(theta)
            + Vt * np.cos(theta)
        )

        mask = r < self.R

        Vx[mask] = np.nan
        Vy[mask] = np.nan

        return Vx, Vy

    # --------------------------------------------------
    # Mouse Press
    # --------------------------------------------------
    def on_press(self, event):

        if event.xdata is None:
            return

        if event.ydata is None:
            return

        distance = np.sqrt(
            (event.xdata - self.x0) ** 2 +
            (event.ydata - self.y0) ** 2
        )

        if distance <= self.R:
            self.dragging = True

    # --------------------------------------------------
    # Mouse Release
    # --------------------------------------------------
    def on_release(self, event):

        self.dragging = False

    # --------------------------------------------------
    # Mouse Drag
    # --------------------------------------------------
    def on_motion(self, event):

        if not self.dragging:
            return

        if event.xdata is None:
            return

        if event.ydata is None:
            return

        # Keep inside plot
        self.x0 = max(-4.5, min(4.5, event.xdata))
        self.y0 = max(-4.5, min(4.5, event.ydata))

        # Update sliders
        self.x_slider.set(round(self.x0, 1))
        self.y_slider.set(round(self.y0, 1))

        self.position_label.config(
            text=f"Center: ({self.x0:.2f}, {self.y0:.2f})"
        )

    # --------------------------------------------------
    # Plot Update
    # --------------------------------------------------
    def update_plot(self, frame):

        self.ax.clear()

        Vx, Vy = self.compute_flow()

        velocity = np.sqrt(
            Vx**2 + Vy**2
        )

        self.ax.streamplot(
            self.X,
            self.Y,
            Vx,
            Vy,
            color=velocity,
            cmap="viridis",
            density=1.7
        )

        cylinder = plt.Circle(
            (self.x0, self.y0),
            self.R,
            fc="dimgray",
            ec="black",
            lw=2,
            zorder=5
        )

        self.ax.add_patch(cylinder)

        self.ax.set_xlim(-5, 5)
        self.ax.set_ylim(-5, 5)

        self.ax.set_aspect(
            "equal",
            adjustable="box"
        )

        self.ax.grid(
            True,
            linestyle="--",
            alpha=0.4
        )

        self.ax.set_xlabel("X")
        self.ax.set_ylabel("Y")

        self.ax.set_title(
            f"Potential Flow Around Cylinder\n"
            f"Radius={self.R:.2f}"
        )

        return []

# ------------------------------------------------------
# Main
# ------------------------------------------------------
if __name__ == "__main__":

    root = tk.Tk()

    app = RealTimeFlowApp(root)

    root.mainloop()
```

```python
#!/usr/bin/env python
# coding: utf-8

"""
Enhanced Flow Over Cylinder Simulation
Improvements over original:
1. Fixed angle calculation using arctan2
2. Added pressure coefficient analysis
3. Enhanced visualization with subplots
4. Added stagnation points marking
5. Included velocity magnitude contours
6. Added theoretical vs actual comparisons
7. Better code organization and documentation
"""

import numpy as np
import matplotlib.pyplot as plt
from numpy import sin, cos, pi

class FlowOverCylinder:
    def __init__(self, U=1.0, R=1.5, rho=1.2, grid_points=300):
        """
        Initialize flow parameters
        
        Parameters:
        U: Freestream velocity (m/s)
        R: Cylinder radius (m)
        rho: Fluid density (kg/m³)
        grid_points: Number of grid points in each direction
        """
        self.U = U
        self.R = R
        self.rho = rho
        self.grid_points = grid_points
        
        # Create computational grid
        self.setup_grid()
        
        # Calculate flow field
        self.calculate_flow_field()
        
    def setup_grid(self):
        """Create computational mesh"""
        x = np.linspace(-5, 5, self.grid_points)
        y = np.linspace(-5, 5, self.grid_points)
        self.X, self.Y = np.meshgrid(x, y)
        
        # Polar coordinates
        self.r = np.sqrt(self.X**2 + self.Y**2)
        self.theta = np.arctan2(self.Y, self.X)  # Fixed: Use arctan2 for all quadrants
        
    def calculate_flow_field(self):
        """Calculate velocity and pressure fields using potential flow theory"""
        # Velocity components in polar coordinates
        self.Vr = self.U * (1 - (self.R**2 / self.r**2)) * cos(self.theta)
        self.Vt = -self.U * (1 + (self.R**2 / self.r**2)) * sin(self.theta)
        
        # Transform to Cartesian coordinates
        self.Vx = self.Vr * cos(self.theta) - self.Vt * sin(self.theta)
        self.Vy = self.Vr * sin(self.theta) + self.Vt * cos(self.theta)
        
        # Velocity magnitude
        self.V_mag = np.sqrt(self.Vx**2 + self.Vy**2)
        
        # Apply boundary condition (inside cylinder)
        mask = self.r < self.R
        self.Vx[mask] = np.nan
        self.Vy[mask] = np.nan
        self.V_mag[mask] = np.nan
        
        # Pressure field using Bernoulli's equation
        self.calculate_pressure()
        
    def calculate_pressure(self):
        """Calculate pressure field and pressure coefficient"""
        # Dynamic pressure
        q_inf = 0.5 * self.rho * self.U**2
        
        # Pressure using Bernoulli's equation: P = P_inf + 0.5*rho*(U_inf^2 - V^2)
        self.P = q_inf * (1 - (self.V_mag / self.U)**2)
        
        # Pressure coefficient: Cp = (P - P_inf) / (0.5 * rho * U_inf^2)
        self.Cp = 1 - (self.V_mag / self.U)**2
        
        # Apply boundary condition
        mask = self.r < self.R
        self.P[mask] = np.nan
        self.Cp[mask] = np.nan
        
    def get_stagnation_points(self):
        """Find stagnation points (theoretical)"""
        stag_points = {
            'front': (-self.R, 0),
            'rear': (self.R, 0)
        }
        return stag_points
        
    def get_surface_pressure(self, n_points=100):
        """Calculate pressure distribution on cylinder surface"""
        theta_surface = np.linspace(0, 2*pi, n_points)
        
        # Theoretical Cp on cylinder surface: Cp = 1 - 4*sin²(θ)
        Cp_surface = 1 - 4 * sin(theta_surface)**2
        
        return theta_surface, Cp_surface
        
    def plot_comprehensive_analysis(self):
        """Create comprehensive flow visualization"""
        fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        fig.suptitle('Flow Over Cylinder - Comprehensive Analysis', fontsize=16, fontweight='bold')
        
        # 1. Streamlines with velocity magnitude
        ax1 = axes[0, 0]
        circle1 = plt.Circle((0, 0), self.R, color='black', fill=True)
        ax1.add_patch(circle1)
        
        strm = ax1.streamplot(self.X, self.Y, self.Vx, self.Vy, 
                             color=self.V_mag, density=[1.5, 1.5], 
                             cmap='viridis', linewidth=1.2)
        
        # Mark stagnation points
        stag_points = self.get_stagnation_points()
        ax1.plot(stag_points['front'][0], stag_points['front'][1], 'ro', markersize=8, label='Stagnation Points')
        ax1.plot(stag_points['rear'][0], stag_points['rear'][1], 'ro', markersize=8)
        
        ax1.set_xlim(-4, 4)
        ax1.set_ylim(-4, 4)
        ax1.set_aspect('equal')
        ax1.set_title('Streamlines & Velocity Magnitude')
        ax1.legend()
        plt.colorbar(strm.lines, ax=ax1, label='Velocity (m/s)')
        
        # 2. Velocity magnitude contours
        ax2 = axes[0, 1]
        circle2 = plt.Circle((0, 0), self.R, color='black', fill=True)
        ax2.add_patch(circle2)
        
        levels = np.linspace(0, 2*self.U, 20)
        cs2 = ax2.contourf(self.X, self.Y, self.V_mag, levels=levels, cmap='plasma')
        ax2.contour(self.X, self.Y, self.V_mag, levels=levels, colors='black', alpha=0.3, linewidths=0.5)
        
        ax2.set_xlim(-4, 4)
        ax2.set_ylim(-4, 4)
        ax2.set_aspect('equal')
        ax2.set_title('Velocity Magnitude Contours')
        plt.colorbar(cs2, ax=ax2, label='Velocity (m/s)')
        
        # 3. Pressure coefficient contours
        ax3 = axes[0, 2]
        circle3 = plt.Circle((0, 0), self.R, color='black', fill=True)
        ax3.add_patch(circle3)
        
        levels_cp = np.linspace(-3, 1, 20)
        cs3 = ax3.contourf(self.X, self.Y, self.Cp, levels=levels_cp, cmap='RdBu_r')
        ax3.contour(self.X, self.Y, self.Cp, levels=levels_cp, colors='black', alpha=0.3, linewidths=0.5)
        
        ax3.set_xlim(-4, 4)
        ax3.set_ylim(-4, 4)
        ax3.set_aspect('equal')
        ax3.set_title('Pressure Coefficient (Cp)')
        plt.colorbar(cs3, ax=ax3, label='Cp')
        
        # 4. Surface pressure distribution
        ax4 = axes[1, 0]
        theta_surf, Cp_surf = self.get_surface_pressure()
        theta_deg = theta_surf * 180 / pi
        
        ax4.plot(theta_deg, Cp_surf, 'b-', linewidth=2, label='Theoretical')
        ax4.axhline(y=0, color='k', linestyle='--', alpha=0.5)
        ax4.axvline(x=90, color='r', linestyle='--', alpha=0.5, label='90° (top)')
        ax4.axvline(x=270, color='r', linestyle='--', alpha=0.5, label='270° (bottom)')
        
        ax4.set_xlabel('Angle θ (degrees)')
        ax4.set_ylabel('Pressure Coefficient (Cp)')
        ax4.set_title('Surface Pressure Distribution')
        ax4.grid(True, alpha=0.3)
        ax4.legend()
        ax4.set_xlim(0, 360)
        
        # 5. Velocity vectors (quiver plot)
        ax5 = axes[1, 1]
        circle5 = plt.Circle((0, 0), self.R, color='black', fill=True)
        ax5.add_patch(circle5)
        
        # Subsample for cleaner vector plot
        skip = 15
        X_sub = self.X[::skip, ::skip]
        Y_sub = self.Y[::skip, ::skip]
        Vx_sub = self.Vx[::skip, ::skip]
        Vy_sub = self.Vy[::skip, ::skip]
        
        ax5.quiver(X_sub, Y_sub, Vx_sub, Vy_sub, 
                  scale=20, alpha=0.7, color='blue', width=0.003)
        
        ax5.set_xlim(-4, 4)
        ax5.set_ylim(-4, 4)
        ax5.set_aspect('equal')
        ax5.set_title('Velocity Vectors')
        
        # 6. Flow parameters and analysis
        ax6 = axes[1, 2]
        ax6.axis('off')
        
        # Calculate some flow metrics
        max_velocity = np.nanmax(self.V_mag)
        min_cp = np.nanmin(self.Cp)
        
        analysis_text = f"""
Flow Parameters:
• Freestream Velocity: {self.U:.1f} m/s
• Cylinder Radius: {self.R:.1f} m
• Fluid Density: {self.rho:.1f} kg/m³

Flow Characteristics:
• Maximum Velocity: {max_velocity:.2f} m/s
• Velocity Ratio (max/U∞): {max_velocity/self.U:.2f}
• Minimum Cp: {min_cp:.2f}

Key Physics:
• Stagnation points at θ = 0°, 180°
• Maximum velocity at θ = 90°, 270°
• Symmetric flow pattern
• Zero net drag (D'Alembert's Paradox)

Theoretical Predictions:
• Max velocity = 2×U∞ at cylinder sides
• Cp = 1 at stagnation points
• Cp = -3 at θ = 90°, 270°
        """
        
        ax6.text(0.05, 0.95, analysis_text, transform=ax6.transAxes, 
                fontsize=10, verticalalignment='top', fontfamily='monospace',
                bbox=dict(boxstyle='round', facecolor='lightgray', alpha=0.8))
        
        plt.tight_layout()
        plt.show()
        
    def plot_comparison_with_theory(self):
        """Compare computed results with theoretical predictions"""
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
        
        # Surface velocity comparison
        theta_surf, Cp_surf = self.get_surface_pressure()
        V_surf_theory = 2 * self.U * np.abs(sin(theta_surf))
        
        ax1.plot(theta_surf * 180/pi, V_surf_theory, 'r-', linewidth=2, label='Theoretical')
        ax1.set_xlabel('Angle θ (degrees)')
        ax1.set_ylabel('Surface Velocity (m/s)')
        ax1.set_title('Surface Velocity Distribution')
        ax1.grid(True, alpha=0.3)
        ax1.legend()
        
        # Pressure coefficient comparison
        ax2.plot(theta_surf * 180/pi, Cp_surf, 'b-', linewidth=2, label='Potential Flow Theory')
        ax2.axhline(y=0, color='k', linestyle='--', alpha=0.5)
        ax2.fill_between(theta_surf * 180/pi, Cp_surf, 0, alpha=0.3)
        ax2.set_xlabel('Angle θ (degrees)')
        ax2.set_ylabel('Pressure Coefficient (Cp)')
        ax2.set_title('Pressure Coefficient Distribution')
        ax2.grid(True, alpha=0.3)
        ax2.legend()
        
        plt.tight_layout()
        plt.show()

# Main execution
if __name__ == "__main__":
    # Create flow simulation
    flow = FlowOverCylinder(U=1.0, R=1.5, rho=1.2, grid_points=250)
    
    # Generate comprehensive analysis
    flow.plot_comprehensive_analysis()
    
    # Generate theoretical comparison
    flow.plot_comparison_with_theory()
    
    print("Flow simulation complete!")
    print(f"Maximum velocity ratio: {np.nanmax(flow.V_mag)/flow.U:.2f}")
    print(f"Minimum pressure coefficient: {np.nanmin(flow.Cp):.2f}")
```
