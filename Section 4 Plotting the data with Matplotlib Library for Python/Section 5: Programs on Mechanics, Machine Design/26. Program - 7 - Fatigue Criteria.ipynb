```python
import numpy as np
import matplotlib.pyplot as plt

# inputs
Sm = 200 # Mean Stress
Sa = 250 # Amplitude Stres

Sut = 450 # Ultimate Stress 
Syt = 330 #Yield Stress
Se = 250 # Endurance Limit  

plt.figure( figsize=(3,3))
           
def soderberg(syt,se):
    # plt.plot([x1,x2],[y1,y2])
    plt.plot([syt,0],[0,se],label= 'Soderberg Line',color='blue')
    plt.xlim(0,Sut+50)
    plt.ylim(0,Se+50)
    plt.legend()
    return

def goodman(sut,se):
    plt.plot([sut,0],[0,se],label= 'Goodman Line',color='red')
    plt.legend()
    return

def gerber(sm,sut,se):
    sm = np.linspace(0,sut,100)
    sa = se*(1 - (sm/sut) ** 2)
    plt.plot(sm, sa, color='green', label='Gerber Curve')
    plt.legend()
    return

(soderberg(Syt,Se) , goodman(Sut,Se) ,gerber(Sm,Sut,Se))

plt.show()

```

```python
import numpy as np
import matplotlib.pyplot as plt

# inputs
Sm = 200 # Mean Stress
Sa = 250 # Amplitude Stres

Sut = 450 # Ultimate Stress 
Syt = 330 #Yield Stress
Se = 250 # Endurance Limit  

plt.figure( figsize=(3,3))
           
def soderberg(syt,se):
    # plt.plot([x1,x2],[y1,y2])
    plt.plot([syt,0],[0,se],label= 'Soderberg Line',color='blue')
    plt.xlim(0,Sut+50)
    plt.ylim(0,Se+50)
    plt.xlabel(r'$\sigma_m$',fontsize = 15)
    plt.ylabel(r'$\sigma_a$',fontsize = 15)
    plt.legend()
    return

def goodman(sut,se):
    plt.plot([sut,0],[0,se],label= 'Goodman Line',color='red')
    plt.legend()
    return

def gerber(sm,sut,se):
    sm = np.linspace(0,sut,100)
    sa = se*(1 - (sm/sut) ** 2)
    plt.plot(sm, sa, color='green', label='Gerber Curve')
    plt.legend()
    return

soderberg(Syt,Se) 
goodman(Sut,Se) 
gerber(Sm,Sut,Se)

smean = [100,200,178,190]
samp = [150,198,230,50]
plt.scatter(smean, samp)

plt.show()

```
