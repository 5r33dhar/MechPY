```python
import tkinter as tk
from tkinter import ttk, messagebox
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg

# ────────────────────────────────────────────────────────────────────────
# CALCULATION & PLOTTING FUNCTION
# ────────────────────────────────────────────────────────────────────────
def sfd_bmd():
    # Try block handles potential conversion errors (e.g., if fields are left blank)
    try:
        # Get values from GUI input fields and convert them to decimals (floats)
        a = float(M_axis.get())
        b = float(m_axis.get())
    except ValueError:
        # If conversion fails, show an error alert popup window and halt execution
        messagebox.showerror("Invalid Input", "Please enter valid numbers in all fields.")
        return

    # Reset/wipe old plots clean from the canvas before drawing a new one
    fig.clear()

    # Create an array of angles from 0 to 360 degrees (in radians) using NumPy
    # Note: Fixed original error by using 'np.pi' instead of undefined 'pi'
    theta = np.arange(0, 2 * np.pi, np.pi / 180)
    
    # Calculate denominator array for the polar coordinates of an ellipse
    # Note: Fixed original errors by using np.sin, np.cos, and np.sqrt
    denom = np.sqrt((a * np.sin(theta)) ** 2 + (b * np.cos(theta)) ** 2)
    
    # Safety Check: Stop calculation if inputs are zero to avoid dividing by zero crashes
    if np.any(denom == 0):
        messagebox.showerror("Calculation Error", "Axes lengths must be greater than zero.")
        return
        
    # Vectorized calculation generating all radial distances (r) instantly
    r = (a * b) / denom

    # Add a clean 1x1 grid subplot specifically configured for polar coordinates
    ax = fig.add_subplot(111, projection='polar')
    
    # Plot the calculated data as red dots ('.' styling format)
    ax.plot(theta, r, 'r.', label='Ellipse Boundary') 
    
    # Add titles and customized underlying styling elements to the graph plot
    ax.set_title("Polar Plot Representation", va='bottom')
    ax.grid(True, linestyle='--', alpha=0.5) # Sets up a subtle grid background

    # Force the embedded Tkinter canvas object to refresh and display the new plot
    canvas.draw()

# ────────────────────────────────────────────────────────────────────────
# CLEAR FIELD DATA FUNCTION
# ────────────────────────────────────────────────────────────────────────
def clear_plot():
    # Remove text strings from position index 0 to the very END inside the entry widgets
    M_axis.delete(0, tk.END)
    m_axis.delete(0, tk.END)
    entry_la.delete(0, tk.END)
    
    # Wipe the figure container clean and force a blank canvas update
    fig.clear()
    canvas.draw()

# ────────────────────────────────────────────────────────────────────────
# MAIN INTERFACE WINDOW STYLING & CORE DESIGN
# ────────────────────────────────────────────────────────────────────────
# Standard initializations setting up main desktop frame layout
beam_win = tk.Tk()
beam_win.title("Polar Plot Geometry Calculator")

# Global variables controlling structural margin gaps around interface elements
pad_x = 10
pad_y = 5

# Major Axis configuration (Row 0)
Major_axis = ttk.Label(beam_win, text="Major axis (a):")
Major_axis.grid(row=0, column=0, padx=pad_x, pady=pad_y, sticky="e")
M_axis = ttk.Entry(beam_win)
M_axis.grid(row=0, column=1, padx=pad_x, pady=pad_y)

# Minor Axis configuration (Row 1)
# Note: Shifted to row=1 to resolve original overlapping collision bug
Minor_axis = ttk.Label(beam_win, text="Minor axis (b):")
Minor_axis.grid(row=1, column=0, padx=pad_x, pady=pad_y, sticky="e")
m_axis = ttk.Entry(beam_win)
m_axis.grid(row=1, column=1, padx=pad_x, pady=pad_y)

# Distance configuration field setup (Row 2)
l_la = ttk.Label(beam_win, text="Distance P (m):")
l_la.grid(row=2, column=0, padx=pad_x, pady=pad_y, sticky="e")
entry_la = ttk.Entry(beam_win)
entry_la.grid(row=2, column=1, padx=pad_x, pady=pad_y)

# Layout Container organizing operations control buttons horizontally (Row 3)
btn_frame = ttk.Frame(beam_win)
btn_frame.grid(row=3, column=0, columnspan=2, pady=10)

# Main action triggering button connected right into calculation function
plot_button = ttk.Button(btn_frame, text="Generate Curve", command=sfd_bmd)
plot_button.grid(row=0, column=0, padx=5)

# Clear/Reset trigger button layout position attachment
clear_button = ttk.Button(btn_frame, text="Clear / Reset", command=clear_plot)
clear_button.grid(row=0, column=1, padx=5)

# Matplotlib Figure Frame mapping setup container size (Row 4)
fig = plt.Figure(figsize=(6, 5), dpi=100)
canvas = FigureCanvasTkAgg(fig, beam_win)
canvas.get_tk_widget().grid(row=4, column=0, columnspan=2, padx=10, pady=10)

# Keeps application continuously scanning for clicks/key strikes until closed
beam_win.mainloop()
```
