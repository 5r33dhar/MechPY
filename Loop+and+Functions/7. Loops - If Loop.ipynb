# IF Statement 

```python
n = -6
if n>0:
    print(n,'is positive number')
else:
    print(n,'is negative number')
```

```python
n = -6
if n>0:
    print(n,'is positive number')
elif n==0:
    print(n,'is zero')
else:
    print(n,'is negative number')
```

```python
n = int(input("Enter number "))
print("Enter number is : ", n)

if n>0:
    print('\n',n,'is positive number')
elif n==0:
    print('\n',n,'is zero')
else:
    print('\n',n,'is negative number')
```

```python
n = 6.5
if n<0:
    x = n ** 2
elif n == 0:
    x = 0
elif n>0:
    x = n ** 0.5
print("The value of x is:",round(x,2))
```

```python
n = -81.69
if n < 6:
    print("x is less than n")
    x = n**2  
elif n == 6:
    print("x is equal to n")
    x = n
else:
    print("x is greater than n")
    x = n**3

print("The value of x is:",round(x,2))
```

```python
for n in range(1,6):
    print("n=",n,"is less than 6")
    if n % 2 > 0:
        print(n,"is odd number")
    else:
        print(n,"is even number")
   
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0,5,50)
print(x,"x values in the range of 0 to 5")

X = []
Y = []
print(X)
print(Y)

for n in x:
    print("n=",n)
    if n <= 3.5:
        y = n**2
    elif n > 3.5 and n <= 4.5:
        y = n**2.5
    else:
        y = n**3
        
    X.append(n)
    Y.append(y)
    
plt.plot(X,Y)
plt.show()
```
