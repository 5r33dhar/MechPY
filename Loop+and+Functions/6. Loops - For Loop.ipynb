```python
sum = 0
n = int(input("Enter number "))
print("Enter number is : ", n)
for i in range(1, n + 1, 1):
    # add current number to sum variable
    sum += i
    print("current number i: ", i, " current Sum: ", sum)
print("\n")
print("final Sum is: ", sum)
```
