```python
n = 5.69
if n < 6:
    print("x is less than n")
    x = n**2  
elif n == 6:
    print("x is equal to n")
    x = n
else:
    print("x is greater than n")
    x = n**3

print("The value of x is:",round(x,2))
    
```

```python
for n in range(1,6):
    if n % 2 > 0:
        print(n,"is odd number")
    else:
        print(n,"is even number")
   
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0,5,50)

X = []
Y = []
for n in x:
    if n <= 2.5:
        y = n**2
    elif n > 2.5:
        y = n**2.5
    X.append(n)
    Y.append(y)
    
plt.plot(X,Y)
plt.show()
```

# While loop

```python
# sum of n numbers

n = 10
n1 = 0
sum = n1
i = 1

while i <=n:
    sum = sum + i
    i = i+1
    
print(sum)
```

# Function

```python
def rectangle_area(b,w):
    return b*w

print(rectangle_area(10,10))
    
```

```python
def four_bar(li,coup,lo):
    if (li+lo) > coup:
        print("This mechanism is possible")
    else:
        print("This is locked mechanism")
        
four_bar(10,18,10)
```
