```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 11)
y = 2 * x ** 3

print('X values \n', x)
print('y values \n', y)

plt.figure(figsize=(4,3))
plt.plot(x, y,color='blue',linestyle='--',lw=3,marker='*')
plt.title("y=F(x)", fontsize=12)
plt.xlabel("X")
plt.ylabel("Y")
plt.xlim(0,10)
plt.ylim(0,200)
plt.show()  # for showing Plot
print('################# \n')

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 11)
y1 = 2 * x ** 2
y2 = 2 * x ** 3

print('X values \n', x)
print('y values \n', y)

print('X last values \n', x[-1])
print('y last values \n', y[-1])

plt.figure(figsize=(4,3))
plt.plot(x, y1,color='red',linestyle='--',lw=1,marker='*',label='y1')
plt.plot(x, y2,color='green',linestyle='-',lw=1,marker='*',label='y2')
plt.legend()

plt.title("y=F(x) last ", fontsize=12)
plt.xlabel('X')
plt.ylabel("Y")
plt.xlim(0,x[-1])
plt.ylim(0,y[-1])
plt.grid()
plt.show()  # for showing Plot
print('################# \n')

import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 11)
y1 = 2 * x ** 2
y2 = 2 * x ** 3

myfig = plt.figure()
#ax1=myfig.add_axes([0,0,1,1])
print('not working \n')
#ax1.plot(x,y1)
plt.show()
print('################# \n')

print('################# \n')

import numpy as np
import matplotlib.pyplot as plt

thetaw  = np.linspace(0, 360, 30)
print('theta values \n', thetaw)


y1 = np.sin(thetaw*np.pi/180)
y2 = np.cos(thetaw*np.pi/180)

x=np.linspace(0.10,11)
print('X values \n', x)
z1=x**2
h3= x ** 3

plt.figure(figsize=(5,5))
plt.subplots(121)
plt.plot(thetaw,y1)
plt.subplots(122)
plt.plot(thetaw,y2)


plt.show()

print('################# \n')

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gsp

x=np.linspace(0.10,11)
print('X values \n', x)
h1=2*x
h2=x**2
h3= x ** 3

fig=plt.figure(figsize= (10,5),dpi=50,facecolor=(1,1,0))

gs=gsp.GridSpec(2,2)
axi1=fig.add_subplot(gs[0,:])
axi2=fig.add_subplot(gs[1,0])
axi3=fig.add_subplot(gs[1,1])

axi1.plot(x,h1)
axi1.set_title('2x')

axi2.plot(x,h2)
axi2.set_title('$x^2$')

axi3.plot(x,h3)
axi3.set_title('$x^3$')


plt.tight_layout()


plt.show()
```
