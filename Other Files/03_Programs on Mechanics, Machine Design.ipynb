```python

import matplotlib.pyplot as plt
import numpy as np
from numpy import pi,sin,cos,sqrt

u0=40 # projectile velocity
g=9.81 #gravity
theta1=30 #projectile angle
theta2=45 #projectile angle

ux1 = u0*cos(theta1*pi/180) # velocity in x 30 degree
uy1 = u0*sin(theta1*pi/180) # velocity in y 30 degree

ux2 = u0*cos(theta2*pi/180) # velocity in x 45 degree
uy2 = u0*sin(theta2*pi/180) # velocity in y 45 degree


t_total1 =2*uy1/g #total time of flight 30 degree
t_total2 =2*uy2/g #total time of flight 45 degree

t1= np.linspace(0,t_total1,100) # time
t2= np.linspace(0,t_total2,100) # time

sx1=ux1*t1
sy1=(uy1*t1)-(0.5*g*t1**2)

sx2=ux2*t2
sy2=(uy2*t2)-(0.5*g*t2**2)

plt.figure(figsize=(5,5))
plt.plot(sx1,sy1,label = r'$\theta={}$'.format(theta1))
plt.plot(sx2,sy2,label ='theta=45')
plt.ylim(0,50)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from numpy import pi,sin,cos,sqrt

#inputs
l=10 # length of beam
W=50 # N/M

#reaction
R=W*l/2

lp=np.linspace(0,l,50) # element as length points
print(lp)

L=[]
SF=[]
MX=[]

for Lp in lp:
    sf=R-(W*Lp) #shear force

    mx=((R*Lp)-(W*Lp*Lp/2))# moment

    L.append(Lp)
    SF.append(sf)
    MX.append(mx)

plt.figure(figsize=(5,5),dpi=100)
plt.subplot(2,1,1)
plt.plot(L,SF,color='green')
plt.fill_between(L,SF,color='yellow',hatch='||',alpha=0.5)

plt.subplot(2,1,2)
plt.plot(L,MX,color='red')
plt.fill_between(L,MX,color='blue',alpha=0.3,hatch='--' )
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from numpy import pi,sin,cos,sqrt


#inputs
L=10 # length of beam
W=50 # N/M

A=2.5
B=5
C=L-(A+B)

#reaction

R1=(W*B/L)*(C+B/2)
R2=(W*B/L)*(A+B/2)

LP=np.linspace(0,L,100)

X=[]
SF=[]
MX=[]

for x in LP:
    if x<A:
        sf=R1
        mx=R1*x

    elif A<x<A+B:
        sf=R1- W*(x-A)
        mx=(R1*x)-W*(x-A)**2/2

    elif  x>A+B:
         sf=-R2
         mx=R2*(L-x)

    X.append(x)
    SF.append(sf)
    MX.append(mx)

plt.figure(figsize=(5,5))

plt.subplot(2,1,1)
plt.plot(X,SF,'g')
plt.fill_between(X,SF,color='g',hatch='//',alpha=0.3)
plt.title('Shear Force Diagram')
plt.xlabel('length of beam')
plt.ylabel('Shear force in N')

plt.tight_layout(pad=2.0)

plt.subplot(2,1,2)
plt.plot(X,MX,'r')
plt.fill_between(X,MX,color='r',hatch='\\',alpha=0.3)
plt.title('Bending Moment  Diagram')
plt.xlabel('length of beam')
plt.ylabel('Bending Moment in N-m')

plt.show()


```

```python
import matplotlib.pyplot as plt
import numpy as np
from numpy import pi, sin, cos, sqrt
from scipy.misc import derivative

r = 125  # crank radius mm
c = 250  # connect rod length mm
rpm = 500  # RPM
omega = (2 * pi * rpm) / 60  # rad/sec


def disp(t):
    d1 = r * cos(omega * t)
    d2 = np.sqrt(c ** 2 - (r ** 2 * sin(omega * t) * sin(omega * t)))
    return d1 + d2


def vel(t):
    return derivative(disp, t)


#def velF(t):
#    derX = r * (omega * t * t) * sin(omega * t) - r ** 2 / 2 * sqrt(c ** 2 - r ** 2 * ((sin(omega * t)) ** 2))
#    return derX


def acc(t):
    return derivative(vel, t)


time = np.linspace(0, 0.5, 300)
theta = omega * time * 180 / pi

x = disp(time)
v = vel(time)
a = acc(time)
# dX=velF(time) ## tried make velocity formal

plt.figure(figsize=(5, 7))
plt.tight_layout(pad=1)

plt.subplot(3, 1, 1)
plt.plot(theta, x, 'r')
plt.title('Displacement ')
plt.xlabel('Degree')
plt.ylabel('Disp')
plt.tight_layout(pad=1)

plt.subplot(3, 1, 2)
plt.plot(theta, v, 'g')
#plt.plot(theta, dX, 'r')
plt.title('Velocity  ')
plt.xlabel('angle of crank Degree')
plt.ylabel('Vel')
plt.tight_layout(pad=1)

plt.subplot(3, 1, 3)
plt.plot(theta, a, 'y')
plt.title('Acceleration ')
plt.xlabel('Degree')
plt.ylabel('Acc')

plt.xlim(0, 360 * 3)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

S_mean =200 # mean stress
S_a=300# amplitude   stress

S_ut=450 # ultimate  stress
S_yt=300 # yield  stress
Se=225 # endurance   stress

plt.figure(figsize=(7,7))

def Soderberg(syt,se):
    # plt.plot ([x1,y1],[x2,y2])
    plt.plot([syt,0],[0,se],label='Soderberg ',color='r')
#   wrong input # plt.plot([0,se],[syt,0] ,label='Soderberg ',color='b')
    plt.xlabel('Stress mean 'r'$\sigma_m$',fontsize=10)
    plt.ylabel('Stress amplitude 'r'$\sigma^a$',fontsize=10)
    plt.xlim(0,S_ut+50)
    plt.ylim(0,se+50)
    return

def Goodman_line (sut,se):
    # plt.plot ([x1,y1],[x2,y2])
    plt.plot([sut,0],[0,se],label='Goodman line ',color='b')
    return

def Gerber_curve (sm,sut,se):
    sm=np.linspace(0,sut,100)
    sa=se*(1-(sm/sut)**2)
    plt.plot(sm,sa,label='Gerber_curve  ',color='y')
    return



Soderberg(S_yt,Se)
Goodman_line(S_ut,Se)
Gerber_curve(S_mean,S_ut,Se)

stress_mean=[100,250,180,50]
stress_amplitude=[150,200,175,50]

plt.scatter(stress_mean,stress_amplitude)
plt.legend()
plt.show()
```
