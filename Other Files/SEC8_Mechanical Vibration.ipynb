```python
# 33. Quarter Car Suspension - Model Equation
# 1/4 of cars suspension
# sprung MASS-CAR BODY Mass
# UNsprung MASS- suspension Mass
# M[a]+C[v]+K[d] = F(t) Sinusoidal road
# a-accleration, v-Velocity ,d-displacement
import matplotlib.pyplot as plt
import numpy as np
from numpy import sin, cos, sqrt, pi

# mass in KG,  K = N/mm  ,c = N-SEC/mm


m1, m2 = 400, 30
k1, k2 = 10000, 40000
#k1, k2 = 10e3, 40e3
c1, c2 = 1000, 0

# force input
a = 0.2  # road bump amplitude mm
v = 0.85  # Velocity  of vehicle m/sec
d = 0.8  # bump with

w = 2 * pi * (v / d)  # frequency of bump

ts = 0
t_step = 0.01
T = np.arange(0, 10, t_step)
print('T',T)

A = np.array([[0.0, 1.0, 0.0, 0.0],
              [-k1 / m1, -c1 / m1, k1 / m1, c1 / m1],
              [0.0, 0.0, 0.0, 1.0],
              [k1 / m2, c1 / m2, -(k1 + k2) / m2, -(c1 + c2) / m2]
              ])

print('A',A)

F = np.array([0.0,0.0,0.0,0.0])
x = np.array([0.0,0.0,0.0,0.0])
print('F',F)

X = []
d1 = []
v1 = []
d2 = []
v2 = []
Ro=[]
for t in T:
    f=(k2 / m2) * a *sin(w*t)
    F[3] = (k2 / m2) * a * sin(w * t)
    x = x + t_step * (A.dot(x) + F)
    #X.append(x)
    d1.append(x[0])
    v1.append(x[1])
    d2.append(x[2])
    v2.append(x[3])
    Ro.append(f)

print(Ro)
plt.figure(figsize=(8, 4), dpi=100)
plt.plot(T, d1, label='displacement of vehicle')
plt.plot(T, d2, label='displacement of wheel')
#plt.plot(T, Ro, label='displacement of Road')
plt.grid(linestyle='--', color='k', alpha=0.8)
plt.legend(loc='best')
plt.xlim(T[0], T[-1])
plt.show()
```
