```python
import tkinter as tk 
# Tkinter library for creating GUI applications

from tkinter import ttk, messagebox 
# Tkinter library for creating GUI applications

import numpy as np 
# Numerical Python library for numerical operations

import matplotlib.pyplot as plt 
# Plotting library for creating visualizations

from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg  
# MATplotlib backend for embedding plots in Tkinter GUI

```

```python
# create gui window
beam_win = tk.Tk() # create a new Tkinter window
beam_win.title("SFD and BMD Calculator: Simply Supported Beam") # set the title of the window


padx = 10 # set the horizontal padding for widgets
pady = 10 # set the vertical padding for widgets

# label for beam length
label_length = tk.Label(beam_win, text="Beam Length BL (m):") # create a label for beam length   
label_length.grid(row=0, column=0, padx=padx, pady=pady) # place the label in the grid layout
Entry_length = tk.Entry(beam_win) # add an entry widget for beam length  to the GUI 
Entry_length.grid(row=0, column=1, padx=padx, pady=pady) # place the entry widget in the grid layout


# label for load location
label_Point_load = tk.Label(beam_win, text="Load Magnitude LM (N):") # Location of the load on the beam   
label_Point_load.grid(row=1, column=0, padx=padx, pady=pady) # place the label load in the grid layout
Point_load = tk.Entry(beam_win) # load ENTRY
Point_load.grid(row=1, column=1, padx=padx, pady=pady) # place the entry widget in the grid layout

# label for Distance of the Point Load From Left Side
label_Distance = tk.Label(beam_win, text="Distance from Left Support LA (m):") # Distance of the point load from the left support
label_Distance.grid(row=2, column=0, padx=padx, pady=pady) # place the label in the grid layout
Entry_Distance = tk.Entry(beam_win) # add an entry widget for distance to the GUI
Entry_Distance.grid(row=2, column=1, padx=padx, pady=pady) # place the entry widget in the grid layout


fig = plt.figure(figsize=(8, 6))  # create a new figure for the shear force and bending moment diagrams
ax1 = fig.add_subplot(211)  # add a subplot for the shear force diagram
canvas = FigureCanvasTkAgg(fig, master=beam_win)  # create a canvas to display the figure in the Tkinter window
canvas.get_tk_widget().grid(row=5, column=0, columnspan=2, padx=padx, pady=pady)  # place the canvas in the grid layout


############################################################################################

def calculate_sfd_bmd():
    lengthBeam = float(Entry_length.get()) # get the beam length from the entry widget L
    load = float(Point_load.get()) # get the load location from the entry widget P
    LengthA= float(Entry_Distance.get()) # get the distance from the entry widget LA

    Ra = load * (1 - (LengthA / lengthBeam))   # reaction at left support
    Rb = load - Ra                # reaction at right support 

   
    ShearForce = []
    Point_lengthBeam = []
    MomentB = []

    for  x in   np.arange (0,lengthBeam+0.01,0.01):  # loop through the length of the beam in increments of 0.01 m
        if x < LengthA:  # if the current position is less than the load location
            shearforce =Ra  # shear force is equal to the reaction at left support
            momentB =Ra*x  # bending moment is equal to reaction at left support times distance from left support
            print(x, shearforce, momentB)
        
        
        else:  # if the current position is greater than the load location
            shearforce =Ra - load  # shear force is equal to reaction at left support minus point load
            momentB =Ra*x - load*(x-LengthA)  # bending moment is equal to reaction at left support times distance from left support minus point load times distance from load location

        ShearForce.append(shearforce)  # append the current shear force to the list of shear forces
        Point_lengthBeam.append(x)  # append the current position to the list of positions along the beam
        MomentB.append(momentB)  # append the current bending moment to the list of bending moments

        print(ShearForce) 
        print(MomentB)
        print(Point_lengthBeam)

        fig.clear()  # clear the current figure
        ax1.step(Point_lengthBeam, ShearForce, where="post", color="blue")
        #ax1.plot(Point_lengthBeam, ShearForce, color='blue')  # plot the shear force diagram
        ax1.set_title('Shear Force Diagram')  # set the title of the shear force diagram
        ax1.set_xlabel('Position along the beam (m)')  # set the x-axis label of the shear force diagram
        ax1.set_ylabel('Shear Force (N)')  # set the y-axis label of the shear force diagram  
        ax1.grid(True)

        return Point_lengthBeam, ShearForce, MomentB  
        # return the lists of positions, shear forces, and bending moments      
          
# Button to calculate SFD and BMD  
plot_button = tk.Button(beam_win, text="Calculate SFD and BMD", command=calculate_sfd_bmd)  

# create a button to calculate SFD and BMD
plot_button.grid(row=4, column=0, columnspan=2, padx=padx, pady=pady)  
# place the button in the grid layout 


# print(ShearForce) 
# print(MomentB)
# print(Point_lengthBeam)          



# ###############################################################################################

canvas.draw()  # draw the canvas to update the plot







# display gui window
beam_win.mainloop() # start the Tkinter event loop
```

```python


# create gui window
beam_win = tk.Tk() # create a new Tkinter window
beam_win.title("SFD and BMD Calculator: Simply Supported Beam") # set the title of the window

padx = 10 # set the horizontal padding for widgets
pady = 10 # set the vertical padding for widgets

# label for beam length
label_length = tk.Label(beam_win, text="Beam Length BL (m):")
label_length.grid(row=0, column=0, padx=padx, pady=pady)
Entry_length = tk.Entry(beam_win)
Entry_length.grid(row=0, column=1, padx=padx, pady=pady)

# label for load magnitude
label_Point_load = tk.Label(beam_win, text="Load Magnitude LM (N):")
label_Point_load.grid(row=1, column=0, padx=padx, pady=pady)
Point_load = tk.Entry(beam_win)
Point_load.grid(row=1, column=1, padx=padx, pady=pady)

# label for distance of the point load from the left support
label_Distance = tk.Label(beam_win, text="Distance from Left Support LA (m):")
label_Distance.grid(row=2, column=0, padx=padx, pady=pady)
Entry_Distance = tk.Entry(beam_win)
Entry_Distance.grid(row=2, column=1, padx=padx, pady=pady)

fig = plt.figure(figsize=(8, 6))
ax1 = fig.add_subplot(211)
ax2 = fig.add_subplot(212)
canvas = FigureCanvasTkAgg(fig, master=beam_win)
canvas.get_tk_widget().grid(row=5, column=0, columnspan=2, padx=padx, pady=pady)


def calculate_sfd_bmd():
    try:
        lengthBeam = float(Entry_length.get())
        load = float(Point_load.get())
        LengthA = float(Entry_Distance.get())
    except ValueError:
        messagebox.showerror("Invalid input", "Enter numeric values for all three inputs.")
        return

    if not np.isfinite([lengthBeam, load, LengthA]).all():
        messagebox.showerror("Invalid input", "All inputs must be finite numbers.")
        return
    if lengthBeam <= 0:
        messagebox.showerror("Invalid input", "Beam length must be greater than zero.")
        return
    if not 0 <= LengthA <= lengthBeam:
        messagebox.showerror("Invalid input", "Load distance must be between 0 and the beam length.")
        return

    Ra = load * (1 - LengthA / lengthBeam)
    Rb = load - Ra

    sample_count = max(2, int(np.ceil(lengthBeam / 0.01)) + 1)
    Point_lengthBeam = np.linspace(0, lengthBeam, sample_count)
    ShearForce = np.where(Point_lengthBeam < LengthA, Ra, Ra - load)
    MomentB = Ra * Point_lengthBeam - load * np.maximum(Point_lengthBeam - LengthA, 0)

    ax1.clear()
    ax1.step(Point_lengthBeam, ShearForce, where="post", color="blue")
    ax1.set_title("Shear Force Diagram")
    ax1.set_ylabel("Shear Force (N)")
    ax1.grid(True)

    ax2.clear()
    ax2.plot(Point_lengthBeam, MomentB, color="red")
    ax2.set_title("Bending Moment Diagram")
    ax2.set_xlabel("Position along the beam (m)")
    ax2.set_ylabel("Bending Moment (N m)")
    ax2.grid(True)

    fig.tight_layout()
    canvas.draw()
    return Point_lengthBeam, ShearForce, MomentB


plot_button = tk.Button(beam_win, text="Calculate SFD and BMD", command=calculate_sfd_bmd)
plot_button.grid(row=4, column=0, columnspan=2, padx=padx, pady=pady)

# display gui window
beam_win.mainloop() # start the Tkinter event loop
```
